###  Data analysis with Pandas

```python
import pandas as pd

data = {"Age": [23, 45, 12, 34]}
df = pd.DataFrame(data)

print(df.head())
print(df.info())
```

### Pandas data-cleaning techniques

```python
import pandas as pd

data = {"Age": [23, 45, 12, 34]}
df = pd.DataFrame(data)

df.dropna(inplace=True)
df.fillna(0, inplace=True)
```

###  Data correlation with Pandas

```python
import pandas as pd

data = {"Age": [23, 45, 12, 34],'Salary': [10,20,30,40]}
df = pd.DataFrame(data)

print(df.corr())
```
