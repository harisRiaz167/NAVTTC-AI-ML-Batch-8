## Assignment 8: Area of Triangle

**Program:**

````python
import math
a, b, c = 5, 6, 7
s = (a + b + c) / 2
area = math.sqrt(s * (s-a) * (s-b) * (s-c))
print("Semi-perimeter =", s)
print("Area =", area)
````

## Assignment 9: Quadratic Equation

**Program:**

````python
import cmath
a, b, c = 1, 5, 6
d = cmath.sqrt(b*b - 4*a*c)
x1 = (-b + d) / (2*a)
x2 = (-b - d) / (2*a)
print(x1, x2)
````

## Assignment 19: Looping Techniques

**Program:**

````python
names = ["Ali", "Sara", "Umer"]
marks = [80, 90, 85]
for i, name in enumerate(names, start=1):
    print(i, name)
for name, mark in zip(names, marks):
    print(name, mark)
student = {"name": "Ali", "marks": 80}
for key, value in student.items():
    print(key, value)
````
