### Assignment 20: User-Defined Functions

**Question:** Write and call user-defined Python functions to perform addition, subtraction, multiplication, and division.


```python

def add(a, b):
    return a + b



def subtract(a, b):
    return a - b



def multiply(a, b):
    return a * b



def divide(a, b):
    if b == 0:
        return "Division by zero is not allowed."
    return a / b



num1 = 20
num2 = 5

print("Addition:", add(num1, num2))
print("Subtraction:", subtract(num1, num2))
print("Multiplication:", multiply(num1, num2))
print("Division:", divide(num1, num2))
```

### Assignment 26: Exception Handling

**Question:** Write a Python program that uses `try`, `except`, `else`, and `finally` to handle invalid numeric input and division by zero.



```python
try:
    numerator = float(input("Enter the numerator: "))
    denominator = float(input("Enter the denominator: "))

    result = numerator / denominator

except ValueError:
    print("Error: Please enter valid numeric values.")

except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")

else:
    print("Result:", result)

finally:
    print("Program execution completed.")
```

### Additional Practice: Multiple Exceptions

**Question:** Write a Python program that converts user input into an integer and handles invalid input and division by zero.

```python
try:
    number = int(input("Enter an integer: "))
    divisor = int(input("Enter a divisor: "))

    result = number / divisor

except ValueError:
    print("Error: You must enter integers only.")

except ZeroDivisionError:
    print("Error: The divisor cannot be zero.")

else:
    print("Division result:", result)

finally:
    print("Execution finished.")
```

### Assignment 27: User-Defined Exceptions

**Question:** Create a custom exception named `InvalidAgeError` and raise it when a user's age is less than 18 or greater than 120.


```python

class InvalidAgeError(Exception):
    pass


def validate_age(age):
    if age < 18 or age > 120:
        raise InvalidAgeError(
            "Age must be between 18 and 120."
        )

    print("Age is valid. You can proceed.")


try:
    age = int(input("Enter your age: "))
    validate_age(age)

except ValueError:
    print("Error: Please enter a valid integer.")

except InvalidAgeError as error:
    print("Invalid age:", error)
```

### Additional Practice: Custom Exception for Marks

**Question:** Create a custom exception named `InvalidMarksError` and raise it when marks are below 0 or above 100.

```python
class InvalidMarksError(Exception):
    pass


def check_marks(marks):
    if marks < 0 or marks > 100:
        raise InvalidMarksError(
            "Marks must be between 0 and 100."
        )

    print("Valid marks:", marks)


try:
    marks = float(input("Enter your marks: "))
    check_marks(marks)

except ValueError:
    print("Error: Enter marks as a number.")

except InvalidMarksError as error:
    print("Invalid marks:", error)
```
