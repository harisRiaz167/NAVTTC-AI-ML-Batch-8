```python
import numpy as np
arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([1.5, 2.5, 3.5])
arr3 = np.array(["Ali", "Sara", "Ahmed"])
print("Integer Array:", arr1)
print("Integer Data Type:", arr1.dtype)
print("Float Array:", arr2)
print("Float Data Type:", arr2.dtype)
print("String Array:", arr3)
print("String Data Type:", arr3.dtype)
arr4 = np.array([1, 2, 3], dtype="float")
print("Float Conversion:", arr4)
```

```python
import numpy as np
original = np.array([10, 20, 30, 40])
copy_arr = original.copy()
view_arr = original.view()
original[0] = 100
print("Original Array:", original)
print("Copy Array:", copy_arr)
print("View Array:", view_arr)
```

```python
import numpy as np
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
arr3 = np.concatenate((arr1, arr2))
print("Joined Array:", arr3)
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
result = np.concatenate((a, b), axis=1)
print("Joined 2D Arrays:")
print(result)
```

```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50, 60])
result = np.array_split(arr, 3)
print("Original Array:", arr)
print("Split Arrays:")
for x in result:
    print(x)
```

```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50, 30])
result = np.where(arr == 30)
print("Positions of 30:", result[0])
result2 = np.where(arr > 25)
print("Positions of Values Greater Than 25:", result2[0])
```

```python
import pandas as pd
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Age": [20, 21, 22, 19],
    "Marks": [85, 90, 78, 95]
}
df = pd.DataFrame(data)
print("Complete DataFrame:")
print(df)
print("First Two Rows:")
print(df.head(2))
print("Student Names:")
print(df["Name"])
print("Students with Marks Greater Than 80:")
print(df[df["Marks"] > 80])
```
