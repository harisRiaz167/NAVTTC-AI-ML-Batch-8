```python
# Complete the exercise on user-defined functions in Python.
# User-Defined Function
def add(a, b):
    result = a + b
    return result
# Taking input from user
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
# Calling the function
answer = add(num1, num2)
print("Sum =", answer)
```

```python
# Complete the exercise on exception handling in Python.
try:
    # Taking input from user
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    # Performing division
    result = num1 / num2
    print("Result =", result)
except ValueError:
    print("Please enter valid integers.")
except ZeroDivisionError:
    print("Error: Second number cannot be zero.")
```

```python
# Complete the exercise on user-defined exception handling in Python.

class CustomError(Exception):
    def __init__(self, message):
        self.message = message

try:
    # Taking input from user
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    
    if num2 == 0:
        raise CustomError("Second number cannot be zero.")
    
    # Performing division
    result = num1 / num2
    print("Result =", result)

except ValueError:
    print("Please enter valid integers.")
except CustomError as e:
    print("Custom Error:", e.message)
```

```python
# Calculate mean, median, and mode.
import statistics
data = [1, 2, 2, 3, 4, 4, 4, 5]
mean = statistics.mean(data)
median = statistics.median(data)
mode = statistics.mode(data)
print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

```python
# Calculate standard deviation.
import statistics
data = [1, 2, 3, 4, 5]
std_dev = statistics.stdev(data)
print("Standard Deviation:", std_dev)
```
