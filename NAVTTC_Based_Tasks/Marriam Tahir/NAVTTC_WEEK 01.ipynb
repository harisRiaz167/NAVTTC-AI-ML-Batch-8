```python
# Find the square root of a real or complex number using Python's cmath module (the example uses 1+2j).
import cmath

# Define a complex number (1 + 2j)
num = 1 + 2j

# Calculate the square root using cmath.sqrt
num_sqrt = cmath.sqrt(num)

# Print the result
print(f"The square root of {num} is {num_sqrt}")

# Format to 3 decimal places for a cleaner layout
print(f"Formatted: {num_sqrt.real:.3f} + {num_sqrt.imag:.3f}j")
```

```python
# find the area of a triangle using its three sides
import math
a = 5
print(f"a = {a}")
b = 6
print(f"b = {b}")
c = 7
print(f"c = {c}")
s = (a + b + c) / 2
area = math.sqrt(s*(s - a) * (s - b) * (s - c))
print(f"Area of Triangle is {area:2f}")
```

```python
# Solve the quadratic equation ax² + bx + c = 0 using the cmath module (the example uses a=1, b=5, c=6).
import cmath

# Define coefficients
a = 1
b = 5
c = 6

# Calculate the discriminant
discriminant = (b**2) - (4*a*c)

# Find two solutions using cmath.sqrt()
sol1 = (-b - cmath.sqrt(discriminant)) / (2*a)
sol2 = (-b + cmath.sqrt(discriminant)) / (2*a)

# Print the results
print(f"The solutions are {sol1} and {sol2}")
```

```python
# Run and examine the provided nested-if example using i = 10 and nested conditions i < 10 and i < 12.

i = 10

if i < 10:          # Outer Condition (Evaluates to False because 10 is not less than 10)
   
   
    print("Outer condition is True")
    if i < 12:      # Inner/Nested Condition (Skipped completely)
        print("Inner condition is True")
else:
    print("The outer condition 'i < 10' is False, so the program skipped the inner check.")
```

```python
# Complete the exercise on various looping techniques in Python.
values = [1, 1, 1, 2, 2, 3]
k = 2
result = []

for value in values:
    if value not in result and values.count(value) >= k:
        result.append(value)
print(result)
```

```python
#  break sttaement

for number in range(1, 6):
    if number == 3:
        print("Found 3! Breaking out of the loop.")
        break
    print(f"Processing number: {number}")

print("Loop ended.")
```

```python
#  User-defined functions in Python
def calc(a: int, b: int, op: int):
    if op == 1:
        return a + b
    elif op == 2:
        return a - b
    elif op == 3:
        return a * b
    elif op == 4:
        if b == 0:
            return "Error: Division by zero"
        return a / b
    else:
        print("Invalid op")
        return None

# Test the function
print(calc(10, 5, 1))  
print(calc(10, 5, 4))  
```

```python
# Complete the exercise on the list data type in Python.
nums = [2, 7, 11, 15]
target = 9
for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i] + nums[j] == target:
            print([i, j])
```
