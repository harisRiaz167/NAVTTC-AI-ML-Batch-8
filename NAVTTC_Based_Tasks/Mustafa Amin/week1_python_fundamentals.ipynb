````python
import importlib
import sys


def check_package(name: str) -> None:
    """Print the installed version of a package, or a hint if it is missing."""
    try:
        module = importlib.import_module(name)
        print(f"{name:<12} {module.__version__}")
    except ImportError:
        print(f"{name:<12} not installed in this environment")


print(f"{'python':<12} {sys.version.split()[0]}")
check_package("torch")
check_package("tensorflow")
````

## Assignment 2 — Linux Commands

These commands are for a Linux terminal, so they are shown as a reference rather than executed
here. Some (`sudo`, `shutdown`, `dpkg`) are Linux-specific and need permissions.

**Navigation and files**
```bash
pwd                         # show current directory
cd /path/to/folder          # change directory
ls -la                      # list files, including hidden ones
cat file.txt                # print a file
less file.txt               # read page by page (q to exit)
more file.txt               # simple page-by-page reading
head -n 5 file.txt          # show first 5 lines
file document.pdf           # identify file type
```

**Creating, copying, and removing**
```bash
mkdir practice              # create directory
touch notes.txt             # create empty file / update timestamp
cp source.txt copy.txt      # copy file
mv old.txt new.txt          # move or rename file
rm file.txt                 # remove a file
rmdir empty_folder          # remove an empty directory
nano notes.txt              # edit a text file
```

**Output redirection**
```bash
echo "hello" > out.txt      # redirect output, overwrite file
echo "again" >> out.txt     # append output
```

**Scripts, processes, and system**
```bash
bash script.sh              # run script with Bash
sh script.sh                # run script with sh
chmod +x script.sh          # make script executable
ps aux                      # list processes
kill PID                    # request a process to stop
sudo command                # run with administrator privileges
man ls                      # manual for ls (q to exit)
alias ll='ls -la'           # define a shell shortcut
dpkg -l                     # list installed Debian packages
sudo shutdown -h now        # power off immediately
sudo systemctl reboot       # restart (systemd-based distributions)
```

**Notes**
- `reading` and `restart` are not standalone Linux commands. Use `cat`/`less` to read files and
  `sudo systemctl reboot` to restart.
- Be careful with `rm` and any privileged command: deletions are not recoverable.

## Assignment 8 — Area of a triangle from three sides

Heron's formula: with $s = (a + b + c) / 2$, the area is $\sqrt{s(s-a)(s-b)(s-c)}$.
The sides must satisfy the triangle inequality, so that is checked first.

````python
import math


def triangle_area(a: float, b: float, c: float) -> float | None:
    """Return the triangle area via Heron's formula, or None if the sides are invalid."""
    if a + b <= c or a + c <= b or b + c <= a:
        return None
    s = (a + b + c) / 2
    return math.sqrt(s * (s - a) * (s - b) * (s - c))


area = triangle_area(5, 6, 7)
print("Area:", area if area is not None else "These sides do not form a triangle.")
````

## Assignment 9 — Solve a quadratic equation

Uses `cmath` so negative discriminants give complex roots instead of raising an error.

````python
import cmath


def solve_quadratic(a: float, b: float, c: float):
    """Return both roots of ax^2 + bx + c = 0, or None if a == 0."""
    if a == 0:
        return None
    discriminant = b**2 - 4 * a * c
    root = cmath.sqrt(discriminant)
    return (-b + root) / (2 * a), (-b - root) / (2 * a)


roots = solve_quadratic(1, 5, 6)
if roots is None:
    print("This is not a quadratic equation because a is 0.")
else:
    print("Root 1:", roots[0])
    print("Root 2:", roots[1])
````

## Assignment 19 — Looping techniques

`enumerate`, `dict.items()`, and `zip`.

````python
names = ["Ali", "Sara", "Hamza"]

# enumerate: index + value
for index, name in enumerate(names, start=1):
    print(index, name)

# dict.items(): key + value
scores = {"Ali": 80, "Sara": 92}
for name, score in scores.items():
    print(f"{name}: {score}")

# zip: iterate over several sequences together
for x, y in zip([1, 2, 3], ["a", "b", "c"]):
    print(x, y)
````

## Assignment 22 — Tuples

Tuples are immutable, so item assignment raises a `TypeError`.

````python
coordinates = (10, 20)
print(coordinates[0])
print("Length:", len(coordinates))

try:
    coordinates[0] = 5
except TypeError as error:
    print("TypeError:", error)
````

## Assignment 24 — Sets

Sets drop duplicates and have no guaranteed order.

````python
numbers = {1, 2, 2, 3, 4}
numbers.add(5)
numbers.discard(2)

print(numbers)
print("Union:", {1, 2, 3} | {3, 4})
print("Intersection:", {1, 2, 3} & {2, 3, 4})
````
