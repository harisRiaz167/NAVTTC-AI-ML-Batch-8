### List size

> The original function added 1 to the length of any non-empty list, so it reported 6 for a list of 5.

```python
def size_of_list(input_list: list) -> dict:
    return {"length": len(input_list)}


print(size_of_list([1, 2, 3, 4, 5]))

assert size_of_list([1, 2, 3, 4, 5]) == {"length": 5}
assert size_of_list([]) == {"length": 0}
```
