## Assignment 28 — Create NumPy arrays

````python
import numpy as np

one_d = np.array([1, 2, 3, 4])
two_d = np.array([[1, 2, 3], [4, 5, 6]])
zeros = np.zeros((2, 3))
ones = np.ones((2, 2))
sequence = np.arange(0, 10, 2)

print("1D:", one_d)
print("2D:\n", two_d)
print("Zeros:\n", zeros)
print("Ones:\n", ones)
print("Sequence:", sequence)
````

## Assignment 29 — Array indexing

````python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("First item:", arr[0])
print("Last item:", arr[-1])
print("Row 2, column 3:", matrix[1, 2])
````

## Assignment 30 — Array slicing

````python
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60])
print("Index 1 to 3:", arr[1:4])
print("From start to index 2:", arr[:3])
print("From index 3 onward:", arr[3:])
print("Every second item:", arr[::2])
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("First column:", matrix[:, 0])
````

## Assignment 31 — NumPy data types

````python
import numpy as np

integers = np.array([1, 2, 3], dtype=np.int32)
decimals = np.array([1.2, 2.5, 3.7], dtype=np.float64)
print(integers, integers.dtype)
print(decimals, decimals.dtype)
converted = integers.astype(float)
print("Converted:", converted, converted.dtype)
````

## Assignment 32 — Copy vs view

````python
import numpy as np

original = np.array([1, 2, 3])
copied = original.copy()
viewed = original.view()

original[0] = 99
print("Original:", original)
print("Copy (independent):", copied)
print("View (shares data):", viewed)
print("View shares memory:", np.shares_memory(original, viewed))
````

## Assignment 33 — Array shape

````python
import numpy as np

arr_1d = np.array([1, 2, 3, 4])
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
arr_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print("1D shape:", arr_1d.shape)
print("2D shape:", arr_2d.shape)
print("3D shape:", arr_3d.shape)
````

## Assignment 34 — Reshape arrays

````python
import numpy as np

arr = np.arange(1, 13)
matrix = arr.reshape(3, 4)
print("Original:", arr)
print("Reshaped to 3x4:\n", matrix)
print("Flattened:", matrix.reshape(-1))
# The requested shape must contain the same number of elements.
````

## Assignment 35 — Iterate through arrays

````python
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])
print("Iterate rows:")
for row in arr:
    print(row)

print("Iterate every element:")
for item in np.nditer(arr):
    print(item)
````

## Assignment 36 — Join arrays

````python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("Concatenate:", np.concatenate((a, b)))

x = np.array([[1, 2], [3, 4]])
y = np.array([[5, 6], [7, 8]])
print("Vertical stack:\n", np.vstack((x, y)))
print("Horizontal stack:\n", np.hstack((x, y)))
````

## Assignment 37 — Split arrays

````python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])
parts = np.array_split(arr, 3)
print("Split into 3 parts:", parts)

matrix = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
print("Split columns:\n", np.array_split(matrix, 2, axis=1))
````

## Assignment 38 — Search arrays

````python
import numpy as np

arr = np.array([2, 4, 6, 4, 8, 4])
print("Indices where value is 4:", np.where(arr == 4)[0])
print("Values greater than 5:", arr[arr > 5])
````

## Assignment 39 — Sort arrays

````python
import numpy as np

arr = np.array([3, 1, 5, 2, 4])
print("Sorted:", np.sort(arr))
matrix = np.array([[3, 2, 1], [6, 5, 4]])
print("Each row sorted:\n", np.sort(matrix, axis=1))
````

## Assignment 40 — NumPy random techniques

````python
import numpy as np

rng = np.random.default_rng(seed=42)  # fixed seed makes results reproducible
print("Random integer:", rng.integers(1, 10))
print("Three random decimals:", rng.random(3))
print("Random 2x3 matrix:\n", rng.integers(0, 100, size=(2, 3)))
print("Choice:", rng.choice(["cat", "dog", "horse"]))
````
