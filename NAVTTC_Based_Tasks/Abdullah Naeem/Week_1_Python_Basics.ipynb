## Assignment 8: Area of a Triangle (Week 1, Day 3)
Find the area of a triangle with sides `a = 5, b = 6, c = 7` using Heron's formula:

- semi-perimeter **s = (a + b + c) / 2**
- area **= √( s(s−a)(s−b)(s−c) )**

```python
a = 5
b = 6
c = 7

s = (a + b + c) / 2
area = (s * (s - a) * (s - b) * (s - c)) ** 0.5

print(f"The area of the triangle is {area:.2f}")
```

## Assignment 9: Solve a Quadratic Equation (Week 1, Day 3)
Solve **ax² + bx + c = 0** for `a = 1, b = 5, c = 6` using `cmath`
(so the program also works when the roots are complex).

```python
a = 1
b = 5
c = 6

discriminant = (b ** 2) - (4 * a * c)

root1 = (-b - cmath.sqrt(discriminant)) / (2 * a)
root2 = (-b + cmath.sqrt(discriminant)) / (2 * a)

print(f"The solutions are {root1} and {root2}")
```

## Assignment 15: `for` Loops (Week 1, Day 5)
A `for` loop goes through the items of a sequence (list, tuple, string, range, dictionary) one by one.

```python
# looping through a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

# looping through a string
for letter in "Python":
    print(letter, end=" ")
print()

# using range(start, stop, step)
for number in range(0, 10, 2):
    print(number, end=" ")
print()

# looping through a dictionary
ages = {"Ali": 21, "Sara": 23}
for name, age in ages.items():
    print(f"{name} is {age} years old")
```

```python
# nested for loop: multiplication table
for row in range(1, 4):
    for col in range(1, 4):
        print(row * col, end="\t")
    print()

# for loop with else: the else block runs when the loop finishes normally
for number in range(3):
    print(number)
else:
    print("Loop finished")
```

## Assignment 16: `while` Loops (Week 1, Day 5)
A `while` loop keeps running as long as its condition is True.

```python
# simple counter
count = 0
while count < 3:
    count = count + 1
    print("Hello Geek")

# while loop with a list
numbers = [1, 2, 3, 4]
while numbers:
    print("Removed:", numbers.pop())

# while loop with else
i = 0
while i < 3:
    print(i)
    i += 1
else:
    print("Condition became False, loop ended")
```

## Assignment 17: `break` Statement (Week 1, Day 5)
`break` stops the loop immediately.

```python
# stop the for loop when the letter 'e' or 's' is found
for letter in "geeksforgeeks":
    if letter == "e" or letter == "s":
        break
    print(letter)
print("Out of for loop, current letter:", letter)

# stop the while loop when i reaches 5
i = 0
while True:
    if i == 5:
        break
    print(i)
    i += 1
print("Out of while loop")
```

## Assignment 18: `continue` Statement (Week 1, Day 5)
`continue` skips the rest of the current iteration and moves to the next one.

```python
# print 1 to 10 but skip 6
for i in range(1, 11):
    if i == 6:
        continue
    print(i, end=" ")
print()

# print a string without the letters 'e' and 's'
for letter in "geeksforgeeks":
    if letter == "e" or letter == "s":
        continue
    print(letter, end="")
print()
```

## Assignment 19: Looping Techniques (Week 1, Day 5)
Useful built-in helpers for looping: `enumerate()`, `zip()`, `.items()`, `sorted()`, and `reversed()`.

```python
names = ["Ali", "Sara", "Usman"]
scores = [88, 92, 75]

# enumerate(): get index and value
for index, name in enumerate(names):
    print(index, name)

# zip(): loop over two lists together
for name, score in zip(names, scores):
    print(f"{name} scored {score}")

# items(): loop over dictionary keys and values
student = {"name": "Ali", "age": 21}
for key, value in student.items():
    print(key, "->", value)

# sorted(): loop in sorted order (set removes duplicates)
for number in sorted(set([3, 1, 4, 1, 5, 9, 2, 6])):
    print(number, end=" ")
print()

# reversed(): loop backwards
for number in reversed(range(1, 6)):
    print(number, end=" ")
print()
```

## Assignment 21: List Data Type (Week 1, Day 4)
A list is an **ordered, changeable** collection written with square brackets `[]`.

```python
# create and access
languages = ["Python", "Swift", "C++"]
print(languages)
print("First item:", languages[0])
print("Last item:", languages[-1])
print("Slice [0:2]:", languages[0:2])

# add items
languages.append("Java")
languages.insert(1, "Go")
languages.extend(["Rust", "Kotlin"])
print("After adding:", languages)

# change an item
languages[2] = "JavaScript"
print("After changing:", languages)

# remove items
languages.remove("Go")
del languages[0]
popped = languages.pop()
print("Popped:", popped)
print("After removing:", languages)
```

```python
numbers = [5, 2, 8, 2, 9, 1]

print("Length:", len(numbers))
print("Count of 2:", numbers.count(2))
print("Index of 8:", numbers.index(8))
print("Is 9 in list?", 9 in numbers)

numbers.sort()
print("Sorted:", numbers)
numbers.reverse()
print("Reversed:", numbers)

# loop through a list
for n in numbers:
    print(n, end=" ")
print()

# list comprehension
squares = [n ** 2 for n in range(1, 6)]
print("Squares:", squares)
```

## Assignment 22: Tuple Data Type (Week 1, Day 4)
A tuple is an **ordered, unchangeable** collection written with round brackets `()`.

```python
# create tuples
numbers = (1, 2, 3)
mixed = (1, "Hello", 3.4)
single = ("Python",)      # a comma is needed for a one-item tuple
print(numbers, mixed, single, type(single))

# access and slice
colors = ("red", "green", "blue", "yellow")
print("First:", colors[0])
print("Last:", colors[-1])
print("Slice [1:3]:", colors[1:3])

# tuple packing and unpacking
person = ("Ali", 21, "Lahore")
name, age, city = person
print(name, age, city)

# tuples cannot be changed
try:
    colors[0] = "black"
except TypeError as error:
    print("Error:", error)

# tuple methods
letters = ("a", "p", "p", "l", "e")
print("Count of 'p':", letters.count("p"))
print("Index of 'l':", letters.index("l"))

# loop and membership
for color in colors:
    print(color, end=" ")
print()
print("Is 'green' in colors?", "green" in colors)
```

## Assignment 23: String Data Type (Week 1, Day 3)
A string is a sequence of characters. Strings are **immutable** (cannot be changed in place).

```python
greet = "Hello, Python!"

# access and slice
print("First character:", greet[0])
print("Last character:", greet[-1])
print("Slice [0:5]:", greet[0:5])

# strings are immutable
try:
    greet[0] = "J"
except TypeError as error:
    print("Error:", error)

# join, repeat, compare
first = "Hello"
second = "World"
print(first + " " + second)
print(first * 3)
print("Are they equal?", first == second)

# loop, length, membership
for char in first:
    print(char, end="-")
print()
print("Length:", len(greet))
print("Is 'Python' in greet?", "Python" in greet)
```

## Assignment 24: Set Data Type (Week 1, Day 4)
A set is an **unordered** collection of **unique** items written with curly brackets `{}`.

```python
# create sets (duplicates are removed automatically)
numbers = {1, 2, 2, 3, 3, 3}
print(numbers)

empty_set = set()          # {} would create an empty dictionary
print(type(empty_set))

# add and update
numbers.add(4)
numbers.update([5, 6])
print("After adding:", numbers)

# remove items
numbers.discard(6)         # no error if the item is missing
numbers.remove(5)          # error if the item is missing
print("After removing:", numbers)

print("Is 3 in set?", 3 in numbers)
```

```python
A = {1, 3, 5, 7}
B = {1, 2, 3, 4}

print("Union        A | B :", A | B)
print("Intersection A & B :", A & B)
print("Difference   A - B :", A - B)
print("Symmetric    A ^ B :", A ^ B)

# the same operations with methods
print(A.union(B), A.intersection(B), A.difference(B), A.symmetric_difference(B))
```

## Assignment 25: Dictionary Data Type (Week 1, Day 4)
A dictionary stores **key: value** pairs. Keys must be unique.

```python
# create and access
country_capitals = {
    "Pakistan": "Islamabad",
    "Italy": "Rome",
    "England": "London",
}
print(country_capitals)
print("Capital of Italy:", country_capitals["Italy"])
print("Using get():", country_capitals.get("Japan", "Not found"))

# add and change items
country_capitals["Japan"] = "Tokyo"
country_capitals["Italy"] = "Milan"
print("After adding/changing:", country_capitals)

# remove items
country_capitals.pop("England")
del country_capitals["Japan"]
print("After removing:", country_capitals)
```

```python
student = {"name": "Ali", "age": 21, "course": "AI"}

print("Keys:", list(student.keys()))
print("Values:", list(student.values()))
print("Items:", list(student.items()))
print("Length:", len(student))
print("Is 'age' a key?", "age" in student)

# loop through a dictionary
for key, value in student.items():
    print(f"{key}: {value}")

# dictionary comprehension
squares = {x: x ** 2 for x in range(1, 6)}
print("Squares:", squares)

student.clear()
print("After clear():", student)
```
