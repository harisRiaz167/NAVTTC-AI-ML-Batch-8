# Two Sum

Given a list of numbers and a target, return the indices of the two numbers that add up to it.
A dictionary of already-seen values gives a single pass, O(n), instead of checking every pair.

Standalone notebook: no imports or input prompts needed. Each `assert` raises an error if the solution is wrong, so no error means it works.

```python
def two_sum(nums: list[int], target: int) -> tuple[int, int] | None:
    """Return (earlier_index, later_index) of two numbers summing to target, or None."""
    seen = {}  # value -> index
    for index, value in enumerate(nums):
        needed = target - value
        if needed in seen:
            return seen[needed], index
        seen[value] = index
    return None


def describe_two_sum(nums: list[int], target: int) -> str:
    pair = two_sum(nums, target)
    if pair is None:
        return "No two numbers in the list add up to that target."
    i, j = pair
    return (f"Target {target} found! Index {i} (value {nums[i]}) + "
            f"Index {j} (value {nums[j]}) = {target}")


numbers = [4, 3, 2, 1]
print(describe_two_sum(numbers, 5))
print(describe_two_sum(numbers, 100))

# Interactive version:
# try:
#     print(describe_two_sum(numbers, int(input("Enter the target sum number: "))))
# except ValueError:
#     print("Please enter a valid whole number.")

assert two_sum([4, 3, 2, 1], 5) == (1, 2)
assert two_sum([2, 7, 11, 15], 9) == (0, 1)
assert two_sum([3, 3], 6) == (0, 1)
assert two_sum([1, 2], 100) is None
```
