```python
df.isna().sum()  #count missing values per column
```

```python
print("Mean:", df["Calories"].mean())
print("Median:", df["Calories"].median())
print("Mode:", df["Calories"].mode())
```

```python
#Null Values/Empty Cell:
new_data_frame=df.dropna()
new_data_frame.head(10)  #new dataframe with no empty cell
```

```python
#Fill null values rows with median
df["Calories"] = df["Calories"].fillna(
    df["Calories"].median()
)

print(df.isnull().sum())
```

**Topic: Statistics**

Calculate mean, median, and mode.

Mean - The average value

Median - The mid point value

Mode - The most common value

```python
import numpy as np
from scipy import stats
```

```python
#scores:
scores = [86, 70, 98, 90, 92, 70, 78, 66]

mean = np.mean(scores)
median = np.median(scores)
modeVal = stats.mode(scores)

print("Scores List:", scores)
print("Mean:", mean )
print("Median:", median)
print("Mode:", modeVal.mode )
```

Calculate standard deviation.

```python
#scores:
scores = [86, 70, 98, 90, 92, 70, 78, 66]
standard_deviation = np.std(scores)

print("Scores List:", scores)
print("Standard Deviation:",round(standard_deviation,3))
```
