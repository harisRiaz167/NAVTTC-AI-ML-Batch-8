# Robot Position Estimation

The robot lives on a grid and starts at `(0, 0)`. It can move **up, down, left or right**, one step at a time.
We estimate its position by adding up every move.

## Step 1: Define the moves

Each move changes `x` (left/right) and `y` (up/down) by a number:

| Move | Change in x | Change in y |
|---|---|---|
| up | 0 | +1 |
| down | 0 | -1 |
| left | -1 | 0 |
| right | +1 | 0 |

```python
moves = {
    "up":    (0, 1),
    "down":  (0, -1),
    "left":  (-1, 0),
    "right": (1, 0),
}
```

## Step 2: Give the robot a list of commands

```python
commands = ["up", "up", "right", "right", "right", "down", "left", "up", "up"]
```

## Step 3: Estimate the position after each move

```python
x, y = 0, 0
path = [(x, y)]            # remember every position

for command in commands:
    dx, dy = moves[command]
    x = x + dx
    y = y + dy
    path.append((x, y))
    print(f"{command:5} -> position ({x}, {y})")

print("Final position:", (x, y))
```

## Step 4: Plot the path

```python
import matplotlib.pyplot as plt

xs = [p[0] for p in path]
ys = [p[1] for p in path]

plt.plot(xs, ys, "o-")
plt.plot(xs[0], ys[0], "go", markersize=12, label="Start")
plt.plot(xs[-1], ys[-1], "ro", markersize=12, label="End")
plt.grid(True)
plt.axis("equal")
plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.title("Robot path")
plt.show()
```
