## Assignment 30: NumPy Slicing

**Program:**

```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50, 60])
print(arr[1:4])
print(arr[:3])
print(arr[::2])
```

## Assignment 31: NumPy Data Types

## Assignment 32: NumPy Copy and View

**Program:**

```python
import numpy as np
original = np.array([1, 2, 3])
copy_arr = original.copy()
view_arr = original.view()
original[0] = 99
print(original)
print(copy_arr)
print(view_arr)
```

## Assignment 35: NumPy Iteration

**Program:**

```python
import numpy as np
arr = np.array([[1, 2], [3, 4]])
for row in arr:
    for value in row:
        print(value)
```

## Assignment 36: Joining NumPy Arrays

**Program:**

```python
import numpy as np
a = np.array([1, 2])
b = np.array([3, 4])
print(np.concatenate((a, b)))
print(np.stack((a, b)))
```

## Assignment 37: Splitting NumPy Arrays

**Program:**

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5, 6])
for part in np.array_split(arr, 3):
    print(part)
```
