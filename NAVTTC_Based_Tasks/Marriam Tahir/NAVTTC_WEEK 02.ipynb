```python
# Practice user defined functions
def find_class_average(marks_list):
    if len(marks_list) == 0:
        raise ValueError("The marks list cannot be empty.")

    total_marks = sum(marks_list)
    class_average = total_marks / len(marks_list)
    return class_average


student_marks = [72, 89, 95, 81, 68]
average_marks = find_class_average(student_marks)

print(f"Student marks: {student_marks}")
print(f"Class average: {average_marks:.2f}")
```

```python
# Complete the exercise on exception handling in Python.
try:
    first_value = float(input("Enter the first number: "))
    second_value = float(input("Enter the second number: "))

    division_result = first_value / second_value

except ValueError:
    print("Please enter numbers only.")

except ZeroDivisionError:
    print("The second number cannot be zero.")

else:
    print(f"The answer is {division_result}")

finally:
    print("The calculation is complete.")
```

```python
# Complete the exercise on user-defined exception handling in Python.
class InvalidStudentAge(Exception):
    pass


def check_student_age(age_years):
    if age_years < 0 or age_years > 120:
        raise InvalidStudentAge("Enter an age between 0 and 120.")

    if age_years >= 18:
        return "The student is eligible as an adult."
    else:
        return "The student is under 18."


try:
    student_age = int(input("Enter the student's age: "))
    age_message = check_student_age(student_age)
    print(age_message)

except ValueError:
    print("Age must be entered as a whole number.")

except InvalidStudentAge as error_message:
    print(f"Invalid age: {error_message}")
```

```python
## Mean, Median, and Mode
from statistics import mean, median, multimode

quiz_scores = [65, 72, 72, 80, 85, 90, 90]

average_score = mean(quiz_scores)
middle_score = median(quiz_scores)
most_common_scores = multimode(quiz_scores)

print(f"Quiz scores: {quiz_scores}")
print(f"Mean: {average_score:.2f}")
print(f"Median: {middle_score}")
print(f"Mode(s): {most_common_scores}")
```

```python
# Standard Deviation
daily_temperatures = [21, 23, 22, 25, 24, 20, 26]

# Calculate the mean
temperature_total = 0

for temperature in daily_temperatures:
    temperature_total += temperature

temperature_count = len(daily_temperatures)
average_temperature = temperature_total / temperature_count

# Calculate the sum of squared differences from the mean
squared_difference_total = 0

for temperature in daily_temperatures:
    difference = temperature - average_temperature
    squared_difference_total += difference ** 2

# Population standard deviation: divide by the total number of values
population_variance = squared_difference_total / temperature_count
population_deviation = population_variance ** 0.5

# Sample standard deviation: divide by one less than the number of values
sample_variance = squared_difference_total / (temperature_count - 1)
sample_deviation = sample_variance ** 0.5

print(f"Daily temperatures: {daily_temperatures}")
print(f"Population standard deviation: {population_deviation:.2f}")
print(f"Sample standard deviation: {sample_deviation:.2f}")
```
