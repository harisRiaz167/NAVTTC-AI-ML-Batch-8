# 4. Paragraph word censor

Merges the three original versions. It flags words, replaces them with asterisks, tidies spacing,
and prints the result in a box.

Changes from the original:

- Matching is by **whole word** and ignores case. The original replaced substrings, so "begun" would have become "be***", and "BOMB" was reported but not censored.
- Capitalising only uppercases the first letter. `str.capitalize()` also lowercases the rest of the sentence, which would turn "Lahore" into "lahore".

Standalone notebook: it includes its own imports and runs top to bottom with no input prompts. Each `assert` raises an error if something is wrong, so no error means the section works.

```python
import re
```

```python
FLAGGED_WORDS = ["bomb", "attack", "kill", "gun"]


def censor_paragraph(paragraph: str, flagged_words=FLAGGED_WORDS):
    """Replace flagged words with asterisks. Returns (censored_text, words_found)."""
    censored = paragraph
    found_words = []
    for word in flagged_words:
        pattern = re.compile(rf"\b{re.escape(word)}\b", re.IGNORECASE)
        if pattern.search(censored):
            found_words.append(word)
            censored = pattern.sub("*" * len(word), censored)
    return censored, found_words


def clean_paragraph(paragraph: str) -> str:
    """Trim, collapse repeated spaces, and capitalise the first letter."""
    cleaned = " ".join(paragraph.split())
    return cleaned[:1].upper() + cleaned[1:]


def display_box(final_paragraph: str, found_words: list, width: int = 60) -> None:
    print("PROCESSED PARAGRAPH".center(width, "="))
    print(final_paragraph.center(width))
    if found_words:
        print(f"Censored: {', '.join(found_words)}".upper().rjust(width))
    else:
        print("No flagged words found.".rjust(width))
    print("=" * width)


def process_paragraph(paragraph: str) -> None:
    censored, found = censor_paragraph(paragraph)
    display_box(clean_paragraph(censored), found)


sample = "  the Attack was a   big BOMB threat, and he had a gun.  "
process_paragraph(sample)
process_paragraph("A calm and peaceful day.")

# Interactive version: process_paragraph(input("Write your paragraph: "))

text, found = censor_paragraph(sample)
assert found == ["bomb", "attack", "gun"]
assert "BOMB" not in text and "Attack" not in text
assert censor_paragraph("He has begun.")[0] == "He has begun."
assert clean_paragraph("   hello    world ") == "Hello world"
assert clean_paragraph("hello Lahore") == "Hello Lahore"
```
