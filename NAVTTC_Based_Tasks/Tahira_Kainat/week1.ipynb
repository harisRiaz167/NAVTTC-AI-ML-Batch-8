Complete the exercise on for loops in Python.

```python
L = [1,3,4,5,6]
T = 7
for i in range(len(L)):
    if L[i] == T:
        print(i)
        break
else:
    insert_at = 0
    while insert_at < len(L) and L[insert_at] < T:
        insert_at += 1
    L.insert(insert_at, T)
    print(insert_at)
    print(L)
```

```python
# Complete the exercise on the break statement in Python.
for i in range(1, 11):
    if i == 5:
        break
    print(i)
```

```python
# Complete the exercise on the continue statement in Python.
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

```python
# Complete the exercise on various looping techniques in Python.
# 1. For Loop
print("For Loop:")
for i in range(1, 6):
    print(i)

# 2. While Loop
print("\nWhile Loop:")
i = 1
while i <= 5:
    print(i)
    i += 1

# 3. Break Statement
print("\nBreak Statement:")
for i in range(1, 6):
    if i == 4:
        break
    print(i)

# 4. Continue Statement
print("\nContinue Statement:")
for i in range(1, 6):
    if i == 3:
        continue
    print(i)

# 5. Enumerate
print("\nEnumerate:")
fruits = ["Apple", "Banana", "Mango"]
for index, fruit in enumerate(fruits):
    print(index, fruit)

# 6. Nested Loop
print("\nNested Loop:")
for i in range(1, 4):
    for j in range(1, 3):
        print(i, j)
```

```python
# Complete the exercise on the string data type in Python.
# String Data Type
name = input("Enter your name: ")

print("Your name is:", name)
print("Length of your name:", len(name))
print("Uppercase:", name.upper())
print("Lowercase:", name.lower())
```

```python
# Complete the exercise on the list data type in Python.
# List Data Type
fruits = ["Apple", "Banana", "Mango", "Orange"]
print("Fruits List:", fruits)
print("First Fruit:", fruits[0])
fruits.append("Grapes")
print("After Adding Fruit:", fruits)
fruits.remove("Banana")
print("After Removing Fruit:", fruits)
print("Total Fruits:", len(fruits))
```

```python
# Complete the exercise on the tuple data type in Python.
# Tuple Data Type
fruits = ("Apple", "Banana", "Mango", "Orange")
print("Fruits Tuple:", fruits)
print("First Fruit:", fruits[0])
print("Last Fruit:", fruits[-1])
print("Total Fruits:", len(fruits))
for fruit in fruits:
    print(fruit)
```

```python
# Complete the exercise on the set data type in Python
# Set Data Type
numbers = {10, 20, 30, 40, 20}
print("Original Set:", numbers)
numbers.add(50)
print("After Adding Number:", numbers)
numbers.remove(30)
print("After Removing Number:", numbers)
print("Total Elements:", len(numbers))
```

```python
# Complete the exercise on the dictionary data type in Python
# Dictionary Data Type
student = {
    "name": "Ali",
    "age": 20,
    "marks": 85
}
print("Student Information:", student)
print("Name:", student["name"])
print("Age:", student["age"])
print("Marks:", student["marks"])
student["grade"] = "A"
print("After Adding Grade:", student)
print("Total Items:", len(student))
```
