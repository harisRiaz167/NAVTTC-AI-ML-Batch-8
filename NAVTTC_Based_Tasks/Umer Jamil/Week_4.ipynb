## Assignment 46: Pandas Data Analysis

**Program:**

```python
import pandas as pd
df = pd.DataFrame({"Name": ["Ali", "Sara", "Umer", "Zara"],
                   "Marks": [70, 90, 80, 100]})
print(df.describe())
print("Average:", df["Marks"].mean())
print("Maximum:", df["Marks"].max())
```

## Assignment 47: Pandas Data Cleaning

**Program:**

```python
import pandas as pd
df = pd.DataFrame({"Name": ["Ali", "Sara", "Sara"],
                   "Marks": [80, None, None]})
print(df.isnull().sum())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df = df.drop_duplicates()
print(df)
```

## Assignment 48: Pandas Correlation

**Program:**

```python
import pandas as pd
df = pd.DataFrame({"StudyHours": [1, 2, 3, 4, 5],
                   "Marks": [50, 55, 65, 70, 80]})
print(df.corr(numeric_only=True))
```
