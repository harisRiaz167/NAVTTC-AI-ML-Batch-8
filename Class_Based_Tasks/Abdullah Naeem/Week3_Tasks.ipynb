## Task 1 — Single Number

Given a non-empty array of integers `nums`, every element appears twice except for one. Find that single one.

```python
nums = [4, 1, 2, 1, 2]

counts = {}
for n in nums:
    counts[n] = counts.get(n, 0) + 1

for n in counts:
    if counts[n] == 1:
        print("Single number:", n)
```

## Task 2 — Missing Number

Given an array `nums` containing `n` distinct numbers in the range `[0, n]`, return the only number in the range that is missing.

```python
numbers = [0, 1, 2, 4]

for i in range(len(numbers) + 1):
    if i not in numbers:
        print("Missing number:", i)
        break
```

## Task 3 — Valid Parentheses

Given a string `s` containing just `()`, `{}` and `[]`, determine if the input string is valid. Open brackets must be closed by the same type and in the correct order.

```python
userInput = input("Enter a parenthesis string: ")


def parenthesis(userInput):
    stack = []

    for i in range(len(userInput)):
        if userInput[i] == "(":
            stack.append(")")
        elif userInput[i] == "[":
            stack.append("]")
        elif userInput[i] == "{":
            stack.append("}")
        else:
            if len(stack) == 0:
                return False
            if stack.pop() != userInput[i]:
                return False

    return len(stack) == 0


print(parenthesis(userInput))
```

## Task 4 — Simple Banking System

Allow users to create accounts, deposit and withdraw money, check balances, and view transaction history.

```python
class BankAccount:
    def __init__(self, owner):
        self.owner = owner
        self.balance = 0
        self.history = []

    def deposit(self, amount):
        self.balance += amount
        self.history.append(f"Deposit: +{amount}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Not enough balance")
            return
        self.balance -= amount
        self.history.append(f"Withdraw: -{amount}")

    def check_balance(self):
        print(f"{self.owner}'s balance: {self.balance}")

    def show_history(self):
        print(f"Transaction history for {self.owner}:")
        for item in self.history:
            print(" ", item)


# Example
account = BankAccount("Alice")
account.deposit(100)
account.withdraw(30)
account.deposit(50)

account.check_balance()
account.show_history()
```

## Task 5 — Valid Anagram

Given two strings `s` and `t`, return `True` if `t` is an anagram of `s`, and `False` otherwise.

```python
s = "anagram"
t = "nagaram"

is_anagram = sorted(s) == sorted(t)
print(is_anagram)
```
