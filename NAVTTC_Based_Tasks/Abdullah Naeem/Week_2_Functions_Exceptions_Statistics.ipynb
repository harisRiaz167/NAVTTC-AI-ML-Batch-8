## Assignment 20: User-Defined Functions (Week 2, Day 1)
A function is a reusable block of code created with `def`.

```python
# simple function
def greet():
    print("Hello from a function")

greet()


# function with parameters and a return value
def add(a, b):
    """Return the sum of two numbers."""
    return a + b

print("add(3, 4) =", add(3, 4))


# default argument
def power(base, exponent=2):
    return base ** exponent

print("power(5) =", power(5))
print("power(2, 3) =", power(2, 3))


# keyword arguments
def student_info(name, age):
    print(f"Name: {name}, Age: {age}")

student_info(age=21, name="Sara")
```

```python
# *args: any number of positional arguments
def total(*numbers):
    return sum(numbers)

print("total(1, 2, 3, 4) =", total(1, 2, 3, 4))


# **kwargs: any number of keyword arguments
def show_details(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

show_details(city="Lahore", country="Pakistan")


# returning multiple values
def min_max(values):
    return min(values), max(values)

smallest, largest = min_max([4, 9, 1, 7])
print("min:", smallest, "max:", largest)


# recursive function
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print("factorial(5) =", factorial(5))


# lambda (anonymous) function
square = lambda x: x * x
print("square(6) =", square(6))
```

## Assignment 26: Exception Handling (Week 2, Day 1)
Use `try`, `except`, `else`, and `finally` so errors do not crash the program.

```python
# basic try / except
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: denominator cannot be 0.")

# catching specific exceptions
even_numbers = [2, 4, 6, 8]
try:
    print(even_numbers[5])
except ZeroDivisionError:
    print("Denominator cannot be 0.")
except IndexError:
    print("Index out of bound.")

# try / except / else / finally
def divide(numerator, denominator):
    try:
        result = numerator / denominator
    except ZeroDivisionError:
        print(f"{numerator}/{denominator}: cannot divide by zero")
    else:
        print(f"{numerator}/{denominator} = {result}")
    finally:
        print("This finally block always runs")

divide(10, 2)
divide(10, 0)
```

## Assignment 27: User-Defined Exceptions (Week 2, Day 1)
Create your own exception class by inheriting from `Exception`, then `raise` it when a rule is broken.

```python
class InvalidAgeException(Exception):
    """Raised when the input age is less than 18."""
    pass


def check_voting_age(age):
    if age < 18:
        raise InvalidAgeException(f"Age {age} is not allowed to vote")
    print(f"Age {age}: eligible to vote")


for age in [25, 15, 18]:
    try:
        check_voting_age(age)
    except InvalidAgeException as error:
        print("Exception occurred:", error)
```

## Assignment 49: Mean, Median, and Mode (Week 2, Day 4)
- **Mean**: the average value
- **Median**: the middle value after sorting
- **Mode**: the most common value

```python
speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

print("Mean  :", round(np.mean(speed), 2))
print("Median:", np.median(speed))
print("Mode  :", stats.mode(speed).mode)
```

## Assignment 50: Standard Deviation (Week 2, Day 5)
Standard deviation shows how spread out the values are. A low value means the numbers are close to the mean.
Variance is the standard deviation squared.

```python
speed_close = [86, 87, 88, 86, 87, 85, 86]
speed_spread = [32, 111, 138, 28, 59, 77, 97]

print("Std (close values) :", round(np.std(speed_close), 2))
print("Std (spread values):", round(np.std(speed_spread), 2))

print("Variance (spread values):", round(np.var(speed_spread), 2))
print("Check: std squared       :", round(np.std(speed_spread) ** 2, 2))
```
