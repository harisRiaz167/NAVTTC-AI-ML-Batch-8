**Assignment 32 — Week 3, Day 4**

**Topic: NumPy**

NumPy array copying and viewing.

```python
original_array = np.array([1, 2, 3, 4, 5])
print("Original Array:", original_array)
# VIEW: A view shares data with the original array. Changes affect both.
view_arr = original_array.view()
print("View of Original Array:", view_arr)
view_arr[1] = 32
print("Original Array after modifying its VIEW:(2nd member (2) changed):", original_array)

# COPY: Changes do not affect the original.
original_arr = np.array([1, 2, 3, 4, 5])
copy_arr = original_arr.copy()
copy_arr[0] = 345
print("Original Array after modifying its COPY:", original_arr)
print("The COPY itself changed:", copy_arr)
```

**Assignment 36 — Week 3, Day 4**

**Topic: NumPy**

Joining NumPy matrices/arrays.

```python
num_arr1 = np.array([[1, 2], [3, 4]])
num_arr2 = np.array([[5, 6], [7, 8]])

concat_arr = np.concatenate((num_arr1, num_arr2), axis=1)
print("concatenate arrays along rows:",concat_arr)

# horizontal joining using hstack
hstack = np.hstack((num_arr1, num_arr2))
print("Horizongtal Joining:",hstack)

# vertical joining using vstack
vstack = np.vstack((num_arr1, num_arr2))
print("Vertical Joining:",vstack)
```

**Assignment 38 — Week 3, Day 4**

**Topic: NumPy**

Searching NumPy arrays.

```python
array = np.array([1, 2, 3, 4, 5, 5, 3, 6])
print("Search Array:",array)

search = np.where(array == 3)#return the indexes, where element=3
print("Indexes where element=3:",search)

evenSearch= np.where(array % 2 == 0)
print("Indexes where element=even:",evenSearch)

ans=np.searchsorted(array, 6) #elment 6 is at index 7
print("Indexes where element=6:",ans)
```
