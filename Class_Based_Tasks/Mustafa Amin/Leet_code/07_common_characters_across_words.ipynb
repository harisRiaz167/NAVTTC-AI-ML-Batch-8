# Common characters across words

Return the characters that appear in **every** word, without repeats, in the order they first
appear in the first word.

Standalone notebook: no imports or input prompts needed. Each `assert` raises an error if the solution is wrong, so no error means it works.

```python
def common_characters(words: list[str]) -> str:
    if not words:
        return ""
    candidates = dict.fromkeys(words[0])  # unique chars of first word, order preserved
    return "".join(ch for ch in candidates if all(ch in word for word in words))


print(common_characters(["income", "in", "intentional", "comein"]))

assert common_characters(["income", "in", "intentional", "comein"]) == "in"
assert common_characters(["abc", "xyz"]) == ""
assert common_characters(["hello"]) == "helo"
assert common_characters([]) == ""
```
