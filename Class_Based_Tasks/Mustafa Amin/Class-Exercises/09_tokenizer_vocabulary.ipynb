# 9. Tokenizer vocabulary

> The original cell used `Counter` and `simple_tokenize` without defining or importing them. Both are
> defined here. Special tokens `[PAD]` and `[UNK]` take ids 0 and 1, and words are numbered by frequency.

Standalone notebook: it includes its own imports and runs top to bottom with no input prompts. Each `assert` raises an error if something is wrong, so no error means the section works.

```python
import re
from collections import Counter

import pandas as pd
```

```python
def simple_tokenize(text: str) -> list:
    return re.findall(r"[a-z']+", text.lower())


def build_vocabulary(sentences: list) -> dict:
    counts = Counter(token for text in sentences for token in simple_tokenize(text))
    vocabulary = {"[PAD]": 0, "[UNK]": 1}
    for token, _ in counts.most_common():
        vocabulary[token] = len(vocabulary)
    return vocabulary


def encode(text: str, vocabulary: dict) -> list:
    """Convert text to ids, using [UNK] for unseen words."""
    return [vocabulary.get(token, vocabulary["[UNK]"]) for token in simple_tokenize(text)]


training_sentences = [
    "the film was very enjoyable",
    "the acting was dull",
    "a delightful and moving story",
    "the ending was weak",
]

vocabulary = build_vocabulary(training_sentences)
vocab_df = pd.DataFrame(vocabulary.items(), columns=["token", "id"])
vocab_df.head(12)
```

```python
print(encode("the film was great", vocabulary))

assert vocabulary["[PAD]"] == 0 and vocabulary["[UNK]"] == 1
assert vocabulary["the"] == 2 and vocabulary["was"] == 3      # most frequent words first
assert encode("the zzz", vocabulary) == [2, 1]                 # unseen word -> [UNK]
assert simple_tokenize("Don't STOP!") == ["don't", "stop"]
```
