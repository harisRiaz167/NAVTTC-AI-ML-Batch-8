### Assignment 17 — Week 1, Day 5

Complete the exercise on the break statement in Python.  
Reference: https://www.geeksforgeeks.org/python-break-statement/


```python
# Question 1: Use break to stop a loop
def stop_loop():
    for i in range(1, 11):
        if i == 5:
            break
        print(i)


# Question 2: Find a number using break
def find_number():
    numbers = [10, 20, 30, 40, 50]

    for number in numbers:
        if number == 30:
            print("Number found:", number)
            break


# Question 3: Stop a while loop using break
def stop_while_loop():
    i = 1

    while i <= 10:
        if i == 6:
            break
        print(i)
        i += 1


# Question 4: Take input until the user enters zero
def input_until_zero():
    while True:
        number = int(input("Enter a number (0 to stop): "))

        if number == 0:
            break

        print("You entered:", number)


# Call all functions
stop_loop()
find_number()
stop_while_loop()
input_until_zero()
```

### Assignment 18 — Week 1, Day 5

Topic: Python  
Complete the exercise on the continue statement in Python.  
Reference: https://www.geeksforgeeks.org/python-continue-statement/


```python
# Question 1: Use continue to skip a number
def skip_number():
    for i in range(1, 6):
        if i == 3:
            continue
        print(i)


# Question 2: Skip even numbers
def skip_even_numbers():
    for i in range(1, 11):
        if i % 2 == 0:
            continue
        print(i)


# Question 3: Skip odd numbers
def skip_odd_numbers():
    for i in range(1, 11):
        if i % 2 != 0:
            continue
        print(i)


# Question 4: Use continue with a while loop
def continue_while():
    i = 0
    while i < 5:
        i += 1
        if i == 3:
            continue
        print(i)


# Question 5: Skip a specific character in a string
def skip_character():
    text = "Python"
    for char in text:
        if char == "h":
            continue
        print(char)


# Question 6: Skip negative numbers
def skip_negative_numbers():
    numbers = [10, -5, 20, -3, 30]
    for number in numbers:
        if number < 0:
            continue
        print(number)


# Question 7: Skip numbers divisible by 3
def skip_divisible_by_three():
    for i in range(1, 11):
        if i % 3 == 0:
            continue
        print(i)


# Question 8: Skip empty strings
def skip_empty_strings():
    words = ["Python", "", "Java", "", "C++"]
    for word in words:
        if word == "":
            continue
        print(word)


# Question 9: Skip a specific word
def skip_word():
    words = ["Apple", "Banana", "Mango", "Orange"]
    for word in words:
        if word == "Mango":
            continue
        print(word)


# Question 10: Take user input and skip zero
def skip_zero():
    for i in range(5):
        number = int(input("Enter a number: "))
        if number == 0:
            continue
        print("Number:", number)


# Call all functions
skip_number()
skip_even_numbers()
skip_odd_numbers()
continue_while()
skip_character()
skip_negative_numbers()
skip_divisible_by_three()
skip_empty_strings()
skip_word()
skip_zero()
```

### Assignment 19 — Week 1, Day 5

Topic: Python  
Complete the exercise on various looping techniques in Python.  
Reference: https://www.geeksforgeeks.org/looping-techniques-python/


```python

def using_for_loop():
    numbers = [1, 2, 3, 4, 5]
    for number in numbers:
        print(number)


def using_while_loop():
    i = 1
    while i <= 5:
        print(i)
        i += 1


def using_range():
    for i in range(1, 6):
        print(i)


def using_enumerate():
    fruits = ["Apple", "Banana", "Mango"]
    for index, fruit in enumerate(fruits):
        print(index, fruit)


def using_zip():
    names = ["Ali", "Sara", "Ahmed"]
    marks = [85, 90, 78]
    for name, mark in zip(names, marks):
        print(name, mark)


def using_list_comprehension():
    squares = [i * i for i in range(1, 6)]
    print(squares)


def using_nested_loop():
    for i in range(1, 4):
        for j in range(1, 4):
            print(i, j)


def using_else():
    for i in range(1, 4):
        print(i)
    else:
        print("Loop completed.")


using_for_loop()
using_while_loop()
using_range()
using_enumerate()
using_zip()
using_list_comprehension()
using_nested_loop()
using_else()
```

### Assignment 21 — Week 1, Day 4

Topic: Python  
Complete the exercise on the list data type in Python.  
Reference: https://www.programiz.com/python-programming/list

```python

def create_list():
    fruits = ["Apple", "Banana", "Mango"]
    print(fruits)


def access_elements():
    fruits = ["Apple", "Banana", "Mango"]
    print(fruits[0])
    print(fruits[-1])


def add_element():
    fruits = ["Apple", "Banana"]
    fruits.append("Mango")
    print(fruits)


def insert_element():
    fruits = ["Apple", "Mango"]
    fruits.insert(1, "Banana")
    print(fruits)


def remove_element():
    fruits = ["Apple", "Banana", "Mango"]
    fruits.remove("Banana")
    print(fruits)


def update_element():
    fruits = ["Apple", "Banana", "Mango"]
    fruits[1] = "Orange"
    print(fruits)


def slice_list():
    numbers = [10, 20, 30, 40, 50]
    print(numbers[1:4])


def sort_list():
    numbers = [5, 2, 8, 1, 3]
    numbers.sort()
    print(numbers)


def reverse_list():
    numbers = [1, 2, 3, 4, 5]
    numbers.reverse()
    print(numbers)


def loop_list():
    fruits = ["Apple", "Banana", "Mango"]
    for fruit in fruits:
        print(fruit)


def list_length():
    fruits = ["Apple", "Banana", "Mango"]
    print(len(fruits))


create_list()
access_elements()
add_element()
insert_element()
remove_element()
update_element()
slice_list()
sort_list()
reverse_list()
loop_list()
list_length()
```

Topic: Python  
Complete the exercise on the dictionary data type in Python.  
Reference: https://www.programiz.com/python-programming/dictionary 

```python

def create_dictionary():
    student = {"name": "Ali", "age": 20, "marks": 85}
    print(student)


def access_value():
    student = {"name": "Ali", "age": 20, "marks": 85}
    print(student["name"])
    print(student.get("marks"))


def add_item():
    student = {"name": "Ali", "age": 20}
    student["marks"] = 85
    print(student)


def update_item():
    student = {"name": "Ali", "age": 20}
    student["age"] = 21
    print(student)


def remove_item():
    student = {"name": "Ali", "age": 20, "marks": 85}
    student.pop("age")
    print(student)


def dictionary_keys_values():
    student = {"name": "Ali", "age": 20, "marks": 85}
    print(student.keys())
    print(student.values())
    print(student.items())


def loop_dictionary():
    student = {"name": "Ali", "age": 20, "marks": 85}
    for key, value in student.items():
        print(key, value)


def check_key():
    student = {"name": "Ali", "age": 20}
    print("name" in student)


def dictionary_length():
    student = {"name": "Ali", "age": 20, "marks": 85}
    print(len(student))


create_dictionary()
access_value()
add_item()
update_item()
remove_item()
dictionary_keys_values()
loop_dictionary()
check_key()
dictionary_length()
```
