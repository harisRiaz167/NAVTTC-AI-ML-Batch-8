## Task 1 — Move Zeroes

Given an integer array `nums`, move all `0`s to the end while keeping the relative order of the non-zero elements.

```python
nums = [0, 1, 0, 3, 12]

result = []
zeros = 0

for n in nums:
    if n != 0:
        result.append(n)
    else:
        zeros += 1

result = result + [0] * zeros
print(result)
```

## Task 2 — Palindrome Number

Given an integer `x`, return `True` if `x` is a palindrome, and `False` otherwise.

```python
x = 121

text = str(x)
is_palindrome = text == text[::-1]

print(is_palindrome)
```

## Task 3 — Majority Element

Given an array `nums` of size `n`, return the element that appears more than `n // 2` times.

```python
nums = [2, 2, 1, 1, 1, 2, 2]

counts = {}
for n in nums:
    counts[n] = counts.get(n, 0) + 1

majority = None
for n in counts:
    if counts[n] > len(nums) // 2:
        majority = n

print(majority)
```

## Task 4 — Simple Grading System

Calculate students' marks, assign grades based on criteria, and display their results.

```python
# Each student has a name and a list of marks
students = {
    "Alice": [85, 90, 78],
    "Bob": [55, 60, 48],
    "Carol": [95, 88, 92],
}


def get_grade(average):
    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


print("RESULTS")
print("-" * 30)
for name, marks in students.items():
    average = sum(marks) / len(marks)
    grade = get_grade(average)
    print(f"{name:<8} Average: {average:5.1f}  Grade: {grade}")
```

## Task 5 — Longest Common Prefix

Find the longest common prefix string among an array of strings. If there is no common prefix, return an empty string `""`.

```python
strs = ["flower", "flow", "flight"]

prefix = strs[0]
for word in strs[1:]:
    while not word.startswith(prefix):
        prefix = prefix[:-1]
        if prefix == "":
            break

print(repr(prefix))
```
