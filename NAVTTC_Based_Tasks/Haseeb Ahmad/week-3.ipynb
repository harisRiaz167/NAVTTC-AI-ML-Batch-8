### Creating NumPy arrays

```python
import numpy as np
arr = np.array([1, 2, 3])
print(arr)
```

### NumPy array indexing

```python
import numpy as np
arr = np.array([10, 20, 30])
print(arr[1])
```

### NumPy array slicing

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
print(arr[1:4])
```

###  NumPy data types

```python
import numpy as np
arr = np.array([1, 2, 3], dtype='f')
print(arr.dtype)
```

###  NumPy array copying and viewing

```python
import numpy as np
arr = np.array([1, 2, 3])
c = arr.copy()
v = arr.view()
print(c, v)
```

###  NumPy array shape

```python
import numpy as np
arr = np.array([[1, 2], [3, 4]])
print(arr.shape)
```

###  NumPy array reshaping

```python
import numpy as np
arr = np.array([1, 2, 3, 4])
print(arr.reshape(2, 2))
```

###  NumPy array iteration

```python
import numpy as np
arr = np.array([1, 2, 3])
for x in arr:
    print(x)
```

### Joining NumPy matrices/arrays

```python
import numpy as np
arr1 = np.array([1, 2])
arr2 = np.array([3, 4])
print(np.concatenate((arr1, arr2)))
```

###  Splitting NumPy arrays


```python
import numpy as np
arr = np.array([1, 2, 3, 4])
print(np.array_split(arr, 2))
```

###  Searching NumPy arrays

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 2])
print(np.where(arr == 2))
```

###  Sorting NumPy arrays

```python
import numpy as np
arr = np.array([3, 1, 2])
print(np.sort(arr))
```

###  NumPy random techniques

```python
import numpy as np
print(np.random.randint(100, size=5))
```
