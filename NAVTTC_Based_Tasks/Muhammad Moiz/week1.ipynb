### Assignment 8
**Question:** Calculate the area of a triangle with sides 5, 6, and 7 using Heron's formula.

```python
import math

a, b, c = 5, 6, 7
s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))
print("Semi-perimeter:", s)
print("Triangle area:", area)
```

### Assignment 9
**Question:** Solve `x² + 5x + 6 = 0` using the `cmath` module.

```python
import cmath

a, b, c = 1, 5, 6
discriminant = b**2 - 4*a*c
root1 = (-b + cmath.sqrt(discriminant)) / (2*a)
root2 = (-b - cmath.sqrt(discriminant)) / (2*a)
print("Root 1:", root1)
print("Root 2:", root2)
```

### Assignment 19
**Question:** Practise looping techniques using `enumerate`, `zip`, and dictionary iteration.


```python
names = ["Ali", "Sara", "Moiz"]
scores = [82, 91, 88]

print("Using enumerate:")
for index, name in enumerate(names, start=1):
    print(index, name)

print("\nUsing zip:")
for name, score in zip(names, scores):
    print(name, "scored", score)

print("\nIterating through a dictionary:")
student_scores = dict(zip(names, scores))
for name, score in student_scores.items():
    print(name, ":", score)
```
