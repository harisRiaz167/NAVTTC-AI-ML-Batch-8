```python
def common_prefix(l):
    if not l:
        return ""

    prefix = l[0]

    for word in l[1:]:
        i = 0

        while i < len(prefix) and i < len(word) and prefix[i] == word[i]:
            i += 1

        prefix = prefix[:i]

        if prefix == "":
            return " "

    return prefix


l = ["income", "in", "index"]

print(common_prefix(l))
l=["dog", "cat", "fish"]
print(common_prefix(l))
```

```python
def merge_list(l1, l2):
    Lo = []
    i = 0
    j = 0
    while i < len(l1) and j < len(l2):
        if l1[i] <= l2[j]:
            Lo.append(l1[i])
            i += 1
        else:
            Lo.append(l2[j])
            j += 1
    while i < len(l1):
        Lo.append(l1[i])
        i += 1
    while j < len(l2):
        Lo.append(l2[j])
        j += 1

    return Lo
l1 = [1, 2, 3, 4]
l2 = [1, 3, 4, 5]
print(merge_list(l1, l2))
```

```python
def is_palindrome(l):
    left = 0
    right = len(l) - 1
    while left < right:
        if l[left] != l[right]:
            return False
        left+= 1
        right-= 1
    return True
l = [1, 2, 3, 3, 2, 1]
print(is_palindrome(l))
```

```python
def anagram(l):
    result = []

    for word in l:
        for oword in l:
            if word != oword and sorted(word) == sorted(oword):
                result.append(word)
                break

    return result


l = ["eat", "ate","tea", "bat", "tan", "nat"]

print(anagram(l))
```

```python
import re
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers
from tensorflow.keras import Sequential


# =====================================================
# 1. Training data
# =====================================================

texts = [
    "I love this movie",
    "This movie is amazing",
    "I really love this film",
    "This film is excellent",

    "I hate this movie",
    "This movie is boring",
    "I really hate this film",
    "This film is terrible"
]

labels = np.array([
    1, 1, 1, 1,
    0, 0, 0, 0
])


# =====================================================
# 2. Tokenization function
# =====================================================

def tokenize(text):

    # Convert to lowercase
    text = text.lower()

    # Get words
    words = re.findall(r"[a-z']+", text)

    return words


# =====================================================
# 3. Create vocabulary
# =====================================================

vocab = {
    "[PAD]": 0,
    "[UNK]": 1
}

for text in texts:

    words = tokenize(text)

    for word in words:

        if word not in vocab:

            vocab[word] = len(vocab)


print("Vocabulary:")
for word, index in vocab.items():
    print(word, "->", index)


# =====================================================
# 4. Convert text into numbers
# =====================================================

def encode_text(text, max_length=6):

    words = tokenize(text)

    encoded = []

    for word in words:

        if word in vocab:

            encoded.append(vocab[word])

        else:

            encoded.append(vocab["[UNK]"])


    # ==============================================
    # Padding
    # ==============================================

    if len(encoded) < max_length:

        padding = max_length - len(encoded)

        encoded = encoded + [vocab["[PAD]"]] * padding


    # ==============================================
    # Truncation
    # ==============================================

    elif len(encoded) > max_length:

        encoded = encoded[:max_length]


    return encoded


# =====================================================
# 5. Encode all training text
# =====================================================

X = []

for text in texts:

    encoded = encode_text(text)

    X.append(encoded)


X = np.array(X)


print("\nEncoded training data:")
print(X)

print("\nShape:")
print(X.shape)


# =====================================================
# 6. Test UNK
# =====================================================

test_text = "I love this fantastic movie"

encoded_test = encode_text(test_text)

print("\nTest sentence:")
print(test_text)

print("\nEncoded test sentence:")
print(encoded_test)


# =====================================================
# 7. Build 1D CNN
# =====================================================

vocab_size = len(vocab)

model = Sequential([

    # Convert word IDs into vectors
    layers.Embedding(
        input_dim=vocab_size,
        output_dim=64
    ),

    # Find patterns in neighboring words
    layers.Conv1D(
        filters=32,
        kernel_size=3,
        activation="relu"
    ),

    # Reduce sequence size
    layers.MaxPooling1D(
        pool_size=2
    ),

    # Convert feature maps into vector
    layers.Flatten(),

    # Fully connected layer
    layers.Dense(
        64,
        activation="relu"
    ),

    # Binary classification
    layers.Dense(
        1,
        activation="sigmoid"
    )
])


# =====================================================
# 8. Compile
# =====================================================

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


# =====================================================
# 9. Show model
# =====================================================

model.summary()


# =====================================================
# 10. Train
# =====================================================

model.fit(
    X,
    labels,
    epochs=20
)


# =====================================================
# 11. Predict new sentences
# =====================================================

new_texts = [
    "I love this movie",
    "This movie is terrible",
    "I love this fantastic movie",
    "This film is boring"
]


for text in new_texts:

    # Convert text → numbers
    encoded = encode_text(text)

    # Convert to NumPy array
    input_data = np.array([encoded])

    # Prediction
    prediction = model.predict(input_data, verbose=0)

    score = prediction[0][0]

    print("\nText:", text)
    print("Encoded:", encoded)
    print("Score:", score)

    if score >= 0.5:

        print("Prediction: Positive")

    else:

        print("Prediction: Negative")
```
