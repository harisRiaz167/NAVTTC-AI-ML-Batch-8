### Assignment 46 — Week 4, Day 1

Topic: Pandas  
Complete the exercise on data analysis with Pandas.  
Reference: https://www.w3schools.com/python/pandas_analyzing.asp

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Usman", "Hina", "Bilal", "Zara"],
    "Age": [20, 22, 21, 23, 24, 20, 22, 21],
    "Marks": [85, 90, 78, 92, 88, 75, 95, 82],
    "City": ["Lahore", "Karachi", "Lahore", "Islamabad", "Multan", "Karachi", "Lahore", "Multan"]
}

df = pd.DataFrame(data)
print(df)
print(df.head())
print(df.head(3))
print(df.tail())
print(df.tail(2))
df.info()
print(df.shape)
print(df.columns)
print(df.dtypes)
print(df.describe())
print(df["Marks"].mean())
print(df["Marks"].max())
print(df["Marks"].min())
print(df["Marks"].sum())
print(df["City"].value_counts())
print(df[df["Marks"] > 85])
print(df.sort_values(by="Marks", ascending=False))
print(df.isnull().sum())
print(df["City"].unique())
print(len(df))
```

### Assignment 47 — Week 4, Day 1

Topic: Pandas  
Complete the exercise on Pandas data-cleaning techniques.  
Reference: https://www.w3schools.com/python/pandas_cleaning.asp

```python

import pandas as pd

df = pd.read_csv("workout_data.csv")

print("Original Dataset:")
print(df.to_string())

print("\nMissing values:")
print(df.isnull().sum())

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

df["Calories"] = df["Calories"].fillna(df["Calories"].mean())

df.loc[df["Duration"] == 450, "Duration"] = 45

df.dropna(subset=["Date"], inplace=True)

df.drop_duplicates(inplace=True)

print("\nCleaned Dataset:")
print(df.to_string(index=False))

print("\nMissing values after cleaning:")
print(df.isnull().sum())

df.to_csv("cleaned_workout_data.csv", index=False)
```

Complete the exercise on data correlation with Pandas.

```python

import pandas as pd
df = pd.read_csv("workout_data.csv")

print(df)
print(df.corr(numeric_only=True))
print(df.corr(numeric_only=True)["Calories"])
print(df.corr(numeric_only=True)["Calories"].sort_values(ascending=False))
```
