**Assignment 27 — Week 2, Day 1**

**Topic: Python**

Custom user-defined exception handling in Python.


```python
print("Custom Exception Handling:")
name=input("Enter your Name:")
age=int(input("Enter your age:"))

class underage_error(Exception):
    """Raised when a user is too young for the License."""
    pass

def check_license(name,age):
    if age < 18:
        raise underage_error(f"Access denied: {name}, you must be at least 18 years old to get the license.")
    else:
      print("License Issued!")

try:
    check_license(name,age)
except underage_error as exception:
    print(exception)
```

**Assignment 49 — Week 2, Day 4**

**Topic: Statistics**

Calculate mean, median, and mode.

**Mean** - The average value

**Median** - The mid point value

**Mode** - The most common value


```python
import numpy as np
from scipy import stats
```

```python
#scores:
scores = [86, 70, 98, 90, 92, 70, 78, 66]

mean = np.mean(scores)
median = np.median(scores)
modeVal = stats.mode(scores)

print("Scores List:", scores)
print("Mean:", mean )
print("Median:", median)
print("Mode:", modeVal.mode )
```
