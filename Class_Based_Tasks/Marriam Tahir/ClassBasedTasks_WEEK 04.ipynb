## Top K Frequent Elements

```python
nums = [1, 1, 1, 2, 2, 3]
k = 2

counts = {}

for number in nums:
    if number in counts:
        counts[number] = counts[number] + 1
    else:
        counts[number] = 1

sorted_numbers = sorted(counts, key=counts.get, reverse=True)
answer = sorted_numbers[:k]

print(answer)
```

## Search Insert Position

```python
nums = [1, 3, 5, 6]
target = 2

position = len(nums)

for i in range(len(nums)):
    if nums[i] >= target:
        position = i
        break

print(position)
```

## Merge Sorted Array

```python
nums1 = [1, 2, 3, 0, 0, 0]
nums2 = [2, 5, 6]

m = 3
n = 3

for i in range(n):
    nums1[m + i] = nums2[i]

nums1.sort()

print(nums1)
```

## 3Sum

```python
nums = [-1, 0, 1, 2, -1, -4]
nums.sort()

triplets = []

for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        for k in range(j + 1, len(nums)):
            if nums[i] + nums[j] + nums[k] == 0:
                triplet = [nums[i], nums[j], nums[k]]

                if triplet not in triplets:
                    triplets.append(triplet)

print(triplets)
```
