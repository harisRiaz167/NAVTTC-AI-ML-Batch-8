# Remove duplicates

Return the unique values of a list in sorted order.

> The original code named its variable `list`, which shadows Python's built-in `list` type and
> breaks later calls like `list(range(10))`. It is renamed here.

Standalone notebook: no imports or input prompts needed. Each `assert` raises an error if the solution is wrong, so no error means it works.

```python
def unique_sorted(values: list) -> list:
    return sorted(set(values))


print(unique_sorted([1, 2, 2, 3, 4, 5]))

assert unique_sorted([1, 2, 2, 3, 4, 5]) == [1, 2, 3, 4, 5]
assert unique_sorted([3, 1, 3, 2, 1]) == [1, 2, 3]
assert unique_sorted([]) == []
```
