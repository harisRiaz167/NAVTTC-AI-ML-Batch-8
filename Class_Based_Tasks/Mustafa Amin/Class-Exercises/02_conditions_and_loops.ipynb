## Even or odd

```python
def is_even(number: int) -> bool:
    return number % 2 == 0


for n in (7, 10):
    print(f"{n} is {'even' if is_even(n) else 'odd'}.")

# Interactive version:
# n = int(input("Enter your number: "))
# print("The number is even." if is_even(n) else "The number is odd.")

assert is_even(10) and not is_even(7) and is_even(0)
```

## Loan eligibility

> The original cell had a syntax error (misspelled names, misaligned `else`). It is rewritten as a function.

```python
def is_eligible_for_loan(has_good_credit: bool, has_stable_income: bool) -> bool:
    return has_good_credit and has_stable_income


for credit, income in [(True, True), (True, False), (False, True)]:
    if is_eligible_for_loan(credit, income):
        print("You are eligible for a loan.")
    else:
        print("You are not eligible for a loan.")

assert is_eligible_for_loan(True, True)
assert not is_eligible_for_loan(True, False)
assert not is_eligible_for_loan(False, False)
```

## Password check

> The original `elif` repeated the same condition as the `if`, so a wrong password printed nothing. It is now an `else`.

```python
def check_password(entered: str, correct: str = "python123") -> str:
    if entered == correct:
        return "Access granted."
    return "Try again. Incorrect password."


print(check_password("python123"))
print(check_password("wrong"))

# Interactive version:
# print(check_password(input("Enter password: ")))

assert check_password("python123") == "Access granted."
assert check_password("nope") == "Try again. Incorrect password."
```

## Pass or fail

```python
def grade_scores(scores: list, pass_mark: int = 50) -> list:
    return [(s, "Passed" if s >= pass_mark else "Failed") for s in scores]


for score, result in grade_scores([45, 72, 88, 39, 91]):
    print(f"Score: {score} - {result}")

assert grade_scores([50, 49]) == [(50, "Passed"), (49, "Failed")]
```

## Number guessing game

The three original versions (repeat until correct, hints, limited attempts) are one function with an
optional `max_attempts`. The guesses come from a function, so it plays interactively or from a script.

> The original hints said "Too close" for a low guess and "Too far" for a high one. They now say "Too low" and "Too high".

```python
HINTS = {
    "low": "Too low! Try again.",
    "high": "Too high! Try again.",
    "correct": "Congratulations! You guessed the correct number.",
}


def check_guess(guess: int, secret: int) -> str:
    if guess == secret:
        return "correct"
    return "low" if guess < secret else "high"


def play_guessing_game(secret: int, get_guess, max_attempts: int | None = None) -> bool:
    """Return True if the player guesses the secret, False if attempts run out."""
    attempts = 0
    while max_attempts is None or attempts < max_attempts:
        result = check_guess(get_guess(), secret)
        attempts += 1
        print(HINTS[result])
        if result == "correct":
            return True
    print(f"Sorry, you've used all your chances. The number was {secret}.")
    return False


print("Unlimited attempts:")
scripted = iter([3, 9, 7])
won = play_guessing_game(7, lambda: next(scripted))

print("\nThree attempts only:")
scripted = iter([1, 2, 3])
lost = play_guessing_game(7, lambda: next(scripted), max_attempts=3)

# Interactive version:
# play_guessing_game(7, lambda: int(input("Guess a number between 1 and 10: ")), max_attempts=3)

assert won is True and lost is False
assert [check_guess(g, 7) for g in (3, 7, 9)] == ["low", "correct", "high"]
```
