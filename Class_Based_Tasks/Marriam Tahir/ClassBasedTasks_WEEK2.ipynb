## Move zeros to end of list

```python
# Move Zeros
nums = [0, 1, 0, 3, 12]

new_list = []

for number in nums:
    if number != 0:
        new_list.append(number)

zero_count = len(nums) - len(new_list)

for i in range(zero_count):
    new_list.append(0)

print(new_list)
```

## Palindrome Number

```python
# Palindrome Number
x = int(input("Enter a number: "))
print(x)

original = x
reversed_number = 0

if x < 0:
    print(False)
else:
    while x > 0:
        digit = x % 10
        reversed_number = reversed_number * 10 + digit
        x = x // 10

    if original == reversed_number:
        print(True)
    else:
        print(False)
```

## Student Grading System

```python
 
name = input("Enter student name: ")

mark1 = float(input("Enter mark for subject 1: "))
mark2 = float(input("Enter mark for subject 2: "))
mark3 = float(input("Enter mark for subject 3: "))

total = mark1 + mark2 + mark3
average = total / 3

if average >= 90:
    grade = "A"
elif average >= 80:
    grade = "B"
elif average >= 70:
    grade = "C"
elif average >= 60:
    grade = "D"
else:
    grade = "F"

print("Student:", name)
print("Total marks:", total)
print("Average:{2f}", average)
print("Grade:", grade)
```

##  Logest Common Prefix

```python

words = ["flower", "flow", "flight"]

prefix = words[0]

for word in words:
    while word[:len(prefix)] != prefix:
        prefix = prefix[:-1]

        if prefix == "":
            break

print(prefix)
```
