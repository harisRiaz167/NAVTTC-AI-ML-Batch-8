different looping techniques, including for, while, break, continue, and else

```python
# 1. For loop
print("For Loop:")
for i in range(1, 6):
    print(i)

# 2. While loop
print("While Loop:")
i = 1
while i <= 5:
    print(i)
    i += 1

# 3. Break statement
print("Break Statement:")
for i in range(1, 6):
    if i == 4:
        break
    print(i)

# 4. Continue statement
print("Continue Statement:")
for i in range(1, 6):
    if i == 3:
        continue
    print(i)

# 5. Loop with else
print("Loop with Else:")
for i in range(1, 4):
    print(i)
else:
    print("Loop completed successfully")
```

call functions, use parameters, and return values.

```python
# Function without parameters
def greet():
    print("Hello, welcome to Python!")

# Function with parameters
def add(a, b):
    return a + b

# Function to find the square
def square(number):
    return number * number

# Calling the functions
greet()

result = add(10, 20)
print("Sum:", result)

result = square(5)
print("Square:", result)
```

 list and perform basic list operations

```python
fruits = ["Apple", "Banana", "Mango", "Orange"]

print("Original list:", fruits)

# Access an element
print("First fruit:", fruits[0])

# Add an element
fruits.append("Grapes")
print("After adding:", fruits)

# Remove an element
fruits.remove("Banana")
print("After removing:", fruits)

# Update an element
fruits[0] = "Pineapple"
print("After updating:", fruits)

# Display list length
print("Length of list:", len(fruits))
```

Create a tuple and access its elements.

```python
colors = ("Red", "Green", "Blue", "Yellow")

print("Original tuple:", colors)

# Access an element
print("First color:", colors[0])

# Access the last element
print("Last color:", colors[-1])

# Display tuple length
print("Length of tuple:", len(colors))

# Check whether an element exists
if "Green" in colors:
    print("Green is present in the tuple")

# Count an element
numbers = (1, 2, 3, 2, 4, 2)
print("Count of 2:", numbers.count(2))
```

Create a string and perform common string operations.

```python
text = "Hello Python"

print("Original string:", text)

# Convert to uppercase
print("Uppercase:", text.upper())

# Convert to lowercase
print("Lowercase:", text.lower())

# Find the length
print("Length:", len(text))

# Access a character
print("First character:", text[0])

# Extract a substring
print("Substring:", text[0:5])

# Replace a word
print("After replacement:", text.replace("Python", "World"))

# Check whether a word exists
print("Contains Python:", "Python" in text)
```

Create a set and perform common set operations.

```python
numbers = {1, 2, 3, 4, 5}

print("Original set:", numbers)

# Add an element
numbers.add(6)
print("After adding:", numbers)

# Remove an element
numbers.remove(3)
print("After removing:", numbers)

# Check membership
print("Is 4 present?", 4 in numbers)

# Union of two sets
a = {1, 2, 3}
b = {3, 4, 5}
print("Union:", a | b)

# Intersection of two sets
print("Intersection:", a & b)

# Difference of two sets
print("Difference:", a - b)
```

 Create a dictionary and perform basic dictionary operations

```python
student = {
    "name": "Ali",
    "age": 20,
    "course": "Python"
}

print("Original dictionary:", student)

# Access a value
print("Student name:", student["name"])

# Add a new key-value pair
student["city"] = "Lahore"
print("After adding city:", student)

# Update a value
student["age"] = 21
print("After updating age:", student)

# Remove a key-value pair
student.pop("course")
print("After removing course:", student)

# Display all keys
print("Keys:", list(student.keys()))

# Display all values
print("Values:", list(student.values()))

```
