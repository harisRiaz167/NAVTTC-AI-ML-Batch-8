# Reverse a string

Return the string reversed. Slicing with a step of `-1` does it in one expression.

Standalone notebook: no imports or input prompts needed. Each `assert` raises an error if the solution is wrong, so no error means it works.

```python
def reverse_string(text: str) -> str:
    return text[::-1]


print(reverse_string("Hello"))

assert reverse_string("Hello") == "olleH"
assert reverse_string("") == ""
assert reverse_string("a") == "a"
```
