## Task 1 — User-defined functions

A user-defined function is a reusable block of code created with `def`.

```python
def greet(name):
    print("Hello,", name)

greet("Sara")
greet("Ali")
```

```python
def add_numbers(num1, num2):
    return num1 + num2

result = add_numbers(5, 3)
print("Sum:", result)
```

## Task 2 — Exception handling

Use `try` and `except` to handle errors without abruptly stopping the program.

```python
try:
    num1 = float(input("Enter the first number: "))
    num2 = float(input("Enter the second number: "))
    result = num1 / num2
    print("Result:", result)
except ValueError:
    print("Please enter valid numbers.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

## Task 3 — User-defined exception handling

A custom exception is created by defining a class that inherits from `Exception`.

```python
class InvalidAgeError(Exception):
    pass

try:
    age = int(input("Enter your age: "))

    if age < 0:
        raise InvalidAgeError("Age cannot be negative.")

    print("Your age is:", age)
except ValueError:
    print("Please enter a whole number.")
except InvalidAgeError as error:
    print("Error:", error)
```

## Task 4 — Calculate mean, median, and mode

The example below uses Python's built-in `statistics` module.

```python
import statistics

numbers = [2, 4, 4, 6, 8, 10]

mean_value = statistics.mean(numbers)
median_value = statistics.median(numbers)
mode_value = statistics.mode(numbers)

print("Numbers:", numbers)
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

## Task 5 — Calculate standard deviation

`statistics.stdev()` calculates the sample standard deviation. `statistics.pstdev()` calculates the population standard deviation.

```python
import statistics

numbers = [2, 4, 4, 6, 8, 10]

sample_sd = statistics.stdev(numbers)
population_sd = statistics.pstdev(numbers)

print("Numbers:", numbers)
print("Sample standard deviation:", sample_sd)
print("Population standard deviation:", population_sd)
```
