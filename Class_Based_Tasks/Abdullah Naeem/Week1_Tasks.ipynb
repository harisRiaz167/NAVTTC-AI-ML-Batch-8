## Task 1 — Maximum and Minimum of an Array

Given an integer array `nums`, find and return the maximum and minimum values.

```python
nums = [3, 7, 1, 9, 2, -5, 8]

maximum = nums[0]
minimum = nums[0]

for n in nums:
    if n > maximum:
        maximum = n
    if n < minimum:
        minimum = n

print("Maximum:", maximum)
print("Minimum:", minimum)
```

## Task 2 — Integer to English Words

Convert a non-negative integer into its English words representation.

```python
ones = ["Zero", "One", "Two", "Three", "Four", "Five", "Six",
        "Seven", "Eight", "Nine", "Ten", "Eleven", "Twelve",
        "Thirteen", "Fourteen", "Fifteen", "Sixteen", "Seventeen",
        "Eighteen", "Nineteen"]
tens = ["", "", "Twenty", "Thirty", "Forty", "Fifty",
        "Sixty", "Seventy", "Eighty", "Ninety"]


def number_to_words(num):
    if num < 20:
        return ones[num]
    if num < 100:
        return tens[num // 10] + (" " + ones[num % 10] if num % 10 else "")
    if num < 1000:
        return ones[num // 100] + " Hundred" + (" " + number_to_words(num % 100) if num % 100 else "")
    if num < 1000000:
        return number_to_words(num // 1000) + " Thousand" + (" " + number_to_words(num % 1000) if num % 1000 else "")
    return number_to_words(num // 1000000) + " Million" + (" " + number_to_words(num % 1000000) if num % 1000000 else "")


print(number_to_words(123))
print(number_to_words(1234567))
```

## Task 3 — Simple Billing System

Manage products, calculate the bill, and print a customer invoice.

```python
# Product catalogue: name -> price
catalogue = {
    "Notebook": 5.50,
    "Pen": 1.20,
    "Backpack": 45.00,
    "Eraser": 0.80,
}

TAX_RATE = 0.06        # 6% tax
DISCOUNT_RATE = 0.10   # 10% off if subtotal is over 50


def make_invoice(customer, cart):
    """cart is a list of (product_name, quantity)."""
    print("INVOICE for", customer)
    print("-" * 35)

    subtotal = 0
    for name, qty in cart:
        price = catalogue[name]
        amount = price * qty
        subtotal += amount
        print(f"{name:<12} {qty} x {price:>6.2f} = {amount:>7.2f}")

    print("-" * 35)
    print(f"{'Subtotal':<22}{subtotal:>7.2f}")

    discount = subtotal * DISCOUNT_RATE if subtotal > 50 else 0
    if discount:
        print(f"{'Discount (10%)':<22}{-discount:>7.2f}")

    tax = (subtotal - discount) * TAX_RATE
    print(f"{'Tax (6%)':<22}{tax:>7.2f}")

    total = subtotal - discount + tax
    print("-" * 35)
    print(f"{'TOTAL':<22}{total:>7.2f}")


# Example
cart = [
    ("Notebook", 3),
    ("Pen", 10),
    ("Backpack", 1),
]
make_invoice("Alice Tan", cart)
```

## Task 4 — Fizz Buzz

Return "FizzBuzz" for multiples of 3 and 5, "Fizz" for multiples of 3, "Buzz" for multiples of 5, otherwise the number.

```python
n = 15
answer = []

for i in range(1, n + 1):
    if i % 15 == 0:
        answer.append("FizzBuzz")
    elif i % 3 == 0:
        answer.append("Fizz")
    elif i % 5 == 0:
        answer.append("Buzz")
    else:
        answer.append(str(i))

print(answer)
```

## Task 5 — Two Sum

Return the indices of the two numbers that add up to `target`.

```python
nums = [int(x) for x in input("Enter numbers separated by spaces: ").split()]
target = int(input("Enter the target: "))

result = []
for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i] + nums[j] == target:
            result = [i, j]

print("Indices:", result)
```
