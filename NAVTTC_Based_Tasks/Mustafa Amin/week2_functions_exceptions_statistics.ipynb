## Assignment 26 — Exception handling

The logic is wrapped in a function so it can be tested with several inputs without waiting for
`input()`. The `else` block runs only when no exception occurs, and `finally` always runs.

```python
def divide(numerator: str, denominator: str) -> None:
    """Divide two values given as text, handling bad input and division by zero."""
    try:
        result = float(numerator) / float(denominator)
    except ValueError:
        print("Please enter valid numbers.")
    except ZeroDivisionError:
        print("Cannot divide by zero.")
    else:
        print("Result:", result)
    finally:
        print("Program finished.\n")


divide("10", "4")     # valid
divide("10", "0")     # ZeroDivisionError
divide("ten", "2")    # ValueError

# Interactive version:
# divide(input("Enter numerator: "), input("Enter denominator: "))
```

## Assignment 27 — User-defined exception

```python
class InvalidAgeError(Exception):
    """Raised when an age is outside the accepted range."""


def check_age(raw_age: str) -> None:
    try:
        age = int(raw_age)
        if age < 0 or age > 120:
            raise InvalidAgeError("Age must be between 0 and 120.")
    except ValueError:
        print("Enter a whole number.")
    except InvalidAgeError as error:
        print("Invalid age:", error)
    else:
        print("Age accepted:", age)


check_age("25")      # valid
check_age("150")     # custom exception
check_age("abc")     # ValueError

# Interactive version:
# check_age(input("Enter your age: "))
```

## Assignment 49 — Mean, median, and mode

- **Mean:** the arithmetic average.
- **Median:** the middle value after sorting.
- **Mode:** the most frequent value. There can be more than one, so `multimode()` returns a list.

```python
data = [2, 4, 4, 6, 9]

print("Mean:", mean(data))
print("Median:", median(data))
print("Mode(s):", multimode(data))
```

## Assignment 50 — Standard deviation

Use `stdev()` for a **sample** and `pstdev()` when the data is the **entire population**.

```python
data = [2, 4, 4, 6, 9]

print("Sample standard deviation:", stdev(data))
print("Population standard deviation:", pstdev(data))
```
