All three assignments use the same example dataset, created below. It contains deliberate
missing values (`None`) so the cleaning step has something to fix.

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Ali", "Sara", "Hamza", "Noor", "Bilal"],
    "age": [22, 24, None, 23, 24],
    "city": ["Lahore", "Karachi", "Lahore", None, "Karachi"],
    "score": [78, 91, 85, None, 91],
})
df
```

## Assignment 46 — Data analysis with Pandas

```python
print(df.head())
print("\nShape (rows, columns):", df.shape)
print()
df.info()
```

```python
print(df.describe())
print("\nAverage score:", df["score"].mean())
```

```python
print("Students by city:")
print(df["city"].value_counts())

print("\nHighest score rows:")
print(df[df["score"] == df["score"].max()])
```

## Assignment 47 — Data cleaning

Steps: check for missing values, fill numeric gaps with the column median, label missing cities
as `"Unknown"`, and drop duplicate rows.

The cleaning is done on a copy (`df_clean`) so the original `df` stays available for comparison.

> Choose filling or removal rules based on the dataset and the problem. Don't replace missing
> values automatically without considering what the gap means.

```python
print("Missing values before cleaning:")
print(df.isna().sum())
```

```python
df_clean = df.copy()

# Fill missing numeric values with each column's median
df_clean["age"] = df_clean["age"].fillna(df_clean["age"].median())
df_clean["score"] = df_clean["score"].fillna(df_clean["score"].median())

# Label missing cities
df_clean["city"] = df_clean["city"].fillna("Unknown")

# Remove duplicate rows, if any
df_clean = df_clean.drop_duplicates()

print(df_clean)
print("\nMissing values after cleaning:")
print(df_clean.isna().sum())
```

## Assignment 48 — Correlation

Correlation ranges from **-1 to +1**:

- near **+1**: strong positive linear association
- near **-1**: strong negative linear association
- near **0**: little linear association

Correlation does not prove causation. It is computed on the cleaned data.

```python
numeric_df = df_clean[["age", "score"]]
print(numeric_df.corr())
```
