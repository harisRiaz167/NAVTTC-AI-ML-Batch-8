## NumPy

## Assignment 28 — Create NumPy arrays

````python
a = np.array([1, 2, 3])
b = np.array([[1, 2], [3, 4]])

print(a)
print(b)
print("Zeros:\n", np.zeros((2, 3)))
print("Range:", np.arange(0, 10, 2))
````

## Assignment 29 — Array indexing

````python
a = np.array([10, 20, 30, 40])
print(a[0])     # first element
print(a[-1])    # last element

matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(matrix[1, 2])  # row 1, column 2
````

## Assignment 30 — Array slicing

````python
a = np.array([10, 20, 30, 40, 50])
print(a[1:4])   # [20 30 40]
print(a[:3])    # [10 20 30]
print(a[::2])   # [10 30 50]

matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(matrix[:, 1])  # second column
````

## Assignment 31 — Data types

````python
integers = np.array([1, 2, 3], dtype=np.int32)
decimals = np.array([1.2, 2.5, 3.8], dtype=np.float64)

print(integers.dtype)
print(decimals.dtype)
print(integers.astype(np.float64))
````

## Assignment 32 — Copy vs. view

A copy owns its data; a view shares memory with the original, so changes show through.

````python
original = np.array([1, 2, 3])
copy_array = original.copy()
view_array = original.view()

original[0] = 99
print("Original:", original)
print("Copy:    ", copy_array, " (independent)")
print("View:    ", view_array, " (shares data)")
````

## Assignment 33 — Array shape

````python
a = np.array([1, 2, 3, 4])
b = np.array([[1, 2, 3], [4, 5, 6]])

print(a.shape)   # (4,)
print(b.shape)   # (2, 3)
print(b.ndim)    # 2 dimensions
````

## Assignment 34 — Reshape

The new shape must contain the same total number of elements. Use `-1` to let NumPy infer one dimension.

````python
a = np.arange(1, 7)
print(a.reshape(2, 3))
print(a.reshape(3, -1))
````

## Assignment 35 — Iterate through arrays

````python
a = np.array([[1, 2], [3, 4]])

print("Every element:")
for item in np.nditer(a):
    print(item)

print("Row by row:")
for row in a:
    print(row)
````

## Assignment 36 — Join arrays

````python
a = np.array([1, 2])
b = np.array([3, 4])
print(np.concatenate((a, b)))

x = np.array([[1, 2], [3, 4]])
y = np.array([[5, 6], [7, 8]])
print("Stack rows (axis=0):\n", np.concatenate((x, y), axis=0))
print("Join columns (axis=1):\n", np.concatenate((x, y), axis=1))
````

## Assignment 37 — Split arrays

For equal splits, the array length must be divisible by the number of parts.

````python
a = np.array([1, 2, 3, 4, 5, 6])
left, right = np.split(a, 2)

print(left)
print(right)
````

## Assignment 38 — Search arrays

````python
a = np.array([4, 2, 7, 2, 9])

print(np.where(a == 2))   # indices where the value is 2
print(np.where(a > 5))
````

## Assignment 39 — Sort arrays

`np.sort` returns a sorted copy and leaves the original unchanged.

````python
a = np.array([3, 1, 5, 2, 4])
print(np.sort(a))
print("Original remains:", a)

matrix = np.array([[3, 1], [4, 2]])
print(np.sort(matrix, axis=0))
````

## Assignment 40 — Random numbers

A seed makes the results repeatable.

````python
rng = np.random.default_rng(42)

print(rng.integers(1, 11, size=5))            # integers 1 through 10
print(rng.random(3))                          # floats in [0, 1)
print(rng.choice(["A", "B", "C"], size=4))
````

## Assignment 41 — Pandas basics

````python
data = {
    "name": ["Ali", "Sara", "Hamza"],
    "age": [22, 24, 21],
    "score": [78, 91, 85],
}
df = pd.DataFrame(data)

print(df)
print()
print(df.head())
print()
df.info()
print()
print(df.describe())
````

## Assignment 43 — Series

````python
scores = pd.Series([78, 91, 85], index=["Ali", "Sara", "Hamza"])

print(scores)
print("Sara:", scores["Sara"])
print("Above 80:")
print(scores[scores > 80])
````

## Assignment 44 — DataFrames

````python
df = pd.DataFrame({
    "name": ["Ali", "Sara", "Hamza"],
    "age": [22, 24, 21],
    "score": [78, 91, 85],
})

print(df["name"])                  # one column
print(df.loc[0])                   # row by label
print(df.iloc[0:2])                # first two rows by position
print(df[df["score"] >= 80])       # filter rows
````

## Assignment 45 — Read a CSV file

The cell below creates `students.csv` in the working directory so the example is self-contained.
`read_csv()` needs the file in the working directory, or you can pass its full path.

````python
from pathlib import Path

csv_path = Path("students.csv")
csv_path.write_text("name,age,score\nAli,22,78\nSara,24,91\nHamza,21,85\n")

df = pd.read_csv(csv_path)
print(df.head())
print("Mean score:", df["score"].mean())
````
