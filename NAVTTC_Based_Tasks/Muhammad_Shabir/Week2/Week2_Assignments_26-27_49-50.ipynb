## Assignment 26 — Exception handling

```python
try:
    value = float(input("Enter a number: "))
    print("You entered:", value)
    print("10 divided by your number:", 10 / value)
except ValueError:
    print("Please enter a valid numeric value.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
finally:
    print("Program finished.")
```

## Assignment 27 — User-defined exception

```python
class NegativeAgeError(Exception):
    """Raised when an age is negative."""

def validate_age(age):
    if age < 0:
        raise NegativeAgeError("Age cannot be negative.")
    return f"Age accepted: {age}"

try:
    age = int(input("Enter age: "))
    print(validate_age(age))
except ValueError:
    print("Enter a whole number.")
except NegativeAgeError as error:
    print("Validation error:", error)
```

## Assignment 49 — Mean, median and mode

```python
from statistics import mean, median, mode

data = [10, 20, 20, 30, 40]
print("Data:", data)
print("Mean:", mean(data))
print("Median:", median(data))
print("Mode:", mode(data))
```

## Assignment 50 — Standard deviation

```python
from statistics import pstdev, stdev

data = [10, 12, 14, 16, 18]
print("Data:", data)
print("Population standard deviation:", pstdev(data))
print("Sample standard deviation:", stdev(data))
```
