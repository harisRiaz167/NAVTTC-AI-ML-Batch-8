### Assignment 20 — Week 2, Day 1

Topic: Python  
Complete the exercise on user-defined functions in Python.  
Reference: https://www.geeksforgeeks.org/functions-in-python/

```python
# Question 1: Create a function with no arguments
def welcome():
    print("Welcome to Python")


# Question 2: Create a function with arguments
def greet(name):
    print("Hello,", name)


# Question 3: Create a function that returns the square of a number
def square(number):
    return number * number


# Question 4: Check whether a number is even or odd
def check_even_odd(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"


# Question 5: Find the first digit of a positive number
def first_digit(number):
    while number >= 10:
        number = number // 10
    return number


# Question 6: Print a message N times
def print_python(n):
    for i in range(n):
        print("Python")


# Question 7: Print numbers from 1 to N without a loop
def print_numbers(n):
    if n == 0:
        return
    print_numbers(n - 1)
    print(n)


# Question 8: Calculate the sum of two numbers
def add_numbers(a, b):
    return a + b


# Question 9: Calculate the power of a number
def power(base, exponent):
    return base ** exponent


# Question 10: Generate a Fibonacci series
def fibonacci(n):
    a = 0
    b = 1
    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b
    print()


# Question 11: Use default arguments in a function
def greet_with_message(name, message="Hello"):
    print(message, name)


# Question 12: Use keyword arguments in a function
def student_info(name, age):
    print("Name:", name)
    print("Age:", age)


# Question 13: Accept any number of positional arguments using *args
def calculate_sum(*args):
    total = 0
    for number in args:
        total += number
    return total


# Question 14: Accept any number of keyword arguments using **kwargs
def display_info(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)


# Question 15: Create a nested function
def outer_function():
    def inner_function():
        print("This is the inner function")
    inner_function()


# Call all functions
welcome()
greet("Ahtisham")
print(square(5))
print(check_even_odd(8))
print(first_digit(4567))
print_python(3)
print_numbers(5)
print(add_numbers(10, 20))
print(power(2, 3))
fibonacci(7)
greet_with_message("Ahtisham")
greet_with_message("Ahtisham", "Welcome")
student_info(age=22, name="Ahtisham")
print(calculate_sum(10, 20, 30, 40))
display_info(name="Ahtisham", age=22, language="Python")
outer_function()
```

### Assignment 26 — Week 2, Day 1

Topic: Python  
Complete the exercise on exception handling in Python.  
Reference: https://www.programiz.com/python-programming/exception-handling


```python
# Question 1: Handle an exception using try and except
def handle_exception():
    try:
        number = int(input("Enter a number: "))
        print(number)
    except ValueError:
        print("Invalid input! Please enter an integer.")


# Question 2: Handle division by zero
def divide_numbers():
    try:
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))
        print("Result:", a / b)
    except ZeroDivisionError:
        print("Error! Cannot divide by zero.")
    except ValueError:
        print("Invalid input! Please enter integers.")


# Question 3: Use multiple except blocks
def multiple_exceptions():
    try:
        number = int(input("Enter a number: "))
        print("Result:", 10 / number)
    except ValueError:
        print("Error! Please enter a valid integer.")
    except ZeroDivisionError:
        print("Error! Cannot divide by zero.")


# Question 4: Use the finally block
def use_finally():
    try:
        number = int(input("Enter a number: "))
        print("Number:", number)
    except ValueError:
        print("Invalid input!")
    finally:
        print("Execution completed.")


# Question 5: Raise an exception using raise
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or above.")
    print("Eligible.")


# Question 6: Handle a raised exception
def handle_raised_exception():
    try:
        check_age(16)
    except ValueError as error:
        print("Error:", error)


# Question 7: Create a custom exception
class NegativeNumberError(Exception):
    pass


def check_positive(number):
    if number < 0:
        raise NegativeNumberError("Negative numbers are not allowed.")
    print("Number:", number)


# Question 8: Handle a custom exception
def handle_custom_exception():
    try:
        check_positive(-5)
    except NegativeNumberError as error:
        print("Error:", error)


# Call all functions
handle_exception()
divide_numbers()
multiple_exceptions()
use_finally()

try:
    check_age(16)
except ValueError as error:
    print("Error:", error)

handle_raised_exception()
check_positive(10)
handle_custom_exception()
```

### Assignment 27 — Week 2, Day 1

Topic: Python  
Complete the exercise on user-defined exception handling in Python.  
Reference: https://www.programiz.com/python-programming/user-defined-exception


```python
# Question 1: Create a user-defined exception
class CustomError(Exception):
    pass


# Question 2: Raise and handle a user-defined exception
def check_number(number):
    try:
        if number < 0:
            raise CustomError("Negative number is not allowed.")
        print("Number:", number)
    except CustomError as error:
        print("Error:", error)


# Call the function
check_number(-5)
check_number(10)
```
