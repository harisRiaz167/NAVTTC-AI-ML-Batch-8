## Five numbers: max, min, unique values, and number words

> The original min/max loops used `for n in list`, which loops over the `list` type instead of the
> user's numbers and fails. Ordinals are also corrected ("4rth" is now "4th").

```python
ORDINALS = ["1st", "2nd", "3rd", "4th", "5th"]
NUMBER_WORDS = {1: "one", 2: "two", 3: "three", 4: "four", 5: "five"}


def read_five_numbers(input_fn=input) -> list:
    return [int(input_fn(f"Enter your {ordinal} number: ")) for ordinal in ORDINALS]


def summarize_numbers(numbers: list) -> dict:
    return {
        "numbers": numbers,
        "max": max(numbers),
        "min": min(numbers),
        "unique": sorted(set(numbers)),
    }


def number_to_word(number: int) -> str:
    return NUMBER_WORDS.get(number, "?")


# Interactive version: user_numbers = read_five_numbers()
user_numbers = [3, 7, 3, 5, 9]
summary = summarize_numbers(user_numbers)

print("The numbers are:", summary["numbers"])
print("Maximum:", summary["max"])
print("Minimum:", summary["min"])
print("Unique: ", summary["unique"])
print("\nNumber words:")
for num in user_numbers:
    print(f"{num}: {number_to_word(num)}")

assert summary["max"] == 9 and summary["min"] == 3
assert summary["unique"] == [3, 5, 7, 9]
assert number_to_word(3) == "three" and number_to_word(7) == "?"

answers = iter(["1", "2", "3", "4", "5"])
assert read_five_numbers(lambda prompt: next(answers)) == [1, 2, 3, 4, 5]
```

## Multiply every number by each element

```python
def products_by_multiplier(numbers: list) -> list:
    """For each element, multiply the whole list by it and sum the products."""
    rows = []
    for index, multiplier in enumerate(numbers):
        products = [n * multiplier for n in numbers]
        rows.append((index, multiplier, products, sum(products)))
    return rows


for index, multiplier, products, total in products_by_multiplier([1, 2, 3, 4]):
    print(f"Index {index} (Value {multiplier}): Products {products} -> Sum is {total}")

assert [row[3] for row in products_by_multiplier([1, 2, 3, 4])] == [10, 20, 30, 40]
```
