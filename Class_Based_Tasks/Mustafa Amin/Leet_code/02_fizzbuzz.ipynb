# FizzBuzz

For each number from 1 to `n`: print `Fizz` for multiples of 3, `Buzz` for multiples of 5,
`FizzBuzz` for multiples of both, otherwise the number itself.

> The original loop only handled multiples of 3 and stopped at 19 (`range(1, 20)`).

Standalone notebook: no imports or input prompts needed. Each `assert` raises an error if the solution is wrong, so no error means it works.

```python
def fizz_buzz(n: int) -> list[str]:
    result = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            result.append("FizzBuzz")
        elif i % 3 == 0:
            result.append("Fizz")
        elif i % 5 == 0:
            result.append("Buzz")
        else:
            result.append(str(i))
    return result


for line in fizz_buzz(20):
    print(line)

assert fizz_buzz(5) == ["1", "2", "Fizz", "4", "Buzz"]
assert fizz_buzz(15)[-1] == "FizzBuzz"
assert fizz_buzz(0) == []
```
