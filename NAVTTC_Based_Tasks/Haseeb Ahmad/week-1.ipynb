###  Python program to add two numbers

```python
num1 = 1.5
num2 = 6.3

sum = num1 + num2

print(sum)
```

### Calculate the square root of a number

```python
num = 8

num_sqrt = 8 ** (1/2)

print(f"Square root of 8 is {num_sqrt}")
```

### Square root of a real or complex number

```python
import cmath


num = 1 + 2j


num_sqrt = cmath.sqrt(num)


print(f"The square root of {num} is {num_sqrt}")
```

### Temperature from Celsius to Fahrenheit

```python
celsius = 37.5


fahrenheit = (celsius * 9/5) + 32


print(f"{celsius}C is equal to {fahrenheit}F")
```

### Area of a triangle

```python
a = 5
b = 6
c = 7

s = (a + b + c) / 2

area = (s * (s - a) * (s - b) * (s - c)) ** 0.5

print(f"The semi-perimeter (s) is: {s}")
print(f"The area of the triangle is: {area}")
```

###  Quadratic equation


```python
import cmath

a = 1
b = 5
c = 6

d = (b**2) - (4 * a * c)

sol1 = (-b - cmath.sqrt(d)) / (2 * a)
sol2 = (-b + cmath.sqrt(d)) / (2 * a)

print(f"The discriminant is: {d}")
print(f"The solutions are {sol1} and {sol2}")
```

###  Distance in kilometers

```python
kilometers = float(input("Enter distance in kilometers: "))

miles = kilometers * 0.621371

print(f"{kilometers} kilometers is equal to {miles} miles.") 

```

### If-statement

```python
i = 10

if i > 15:
    print("i is greater than 15")
else:
    print("i is not greater than 15")
```

###  If-else example

```python
i = 20
if i < 15:
    print("i is less than 15")
else:
    print("i is not less than 15")
```

### Nested-if example

```python
i = 10
if i < 12:
    if i < 10:
        print("i is less than 10")
    else:
        print("i is between 10 and 11")
```

###  If-elif-else example

```python
i = 20
if i == 10:
    print("i is 10")
elif i == 15:
    print("i is 15")
elif i == 20:
    print("i is 20")
else:
    print("i is something else")
```

### For loops in Python

```python
for item in range(5):
    print(item)
```

###  While loops in Python

```python
count = 0
while count < 5:
    print(count)
    count += 1
```

###  Break statement in Python

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

### Continue statement in Python

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
```

### Looping techniques in Python

```python
items = ['a', 'b', 'c']
for index, value in enumerate(items):
    print(index, value)
```
