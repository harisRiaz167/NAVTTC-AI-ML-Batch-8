## Task 8 — Area of a triangle using its three sides (Heron's formula)

```python
import math

a = 5
b = 6
c = 7

s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))

print("Semi-perimeter:", s)
print("Area of the triangle:", area)
```

## Task 9 — Solve a quadratic equation using `cmath`

```python
import cmath

a = 1
b = 5
c = 6

discriminant = b**2 - 4*a*c
root1 = (-b + cmath.sqrt(discriminant)) / (2*a)
root2 = (-b - cmath.sqrt(discriminant)) / (2*a)

print("First root:", root1)
print("Second root:", root2)
```

## Task 14 — Set data type

Sets contain unique values and do not guarantee a fixed display order.

```python
numbers = {1, 2, 2, 3, 4, 4}
print("Set (duplicates removed):", numbers)

numbers.add(5)
print("After adding 5:", numbers)

numbers.discard(2)
print("After discarding 2:", numbers)

first = {1, 2, 3}
second = {3, 4, 5}
print("Union:", first | second)
print("Intersection:", first & second)
print("Difference:", first - second)
```

## Task 24 — Various looping techniques

```python
names = ["Ali", "Sara", "Ahmed"]

print("Using enumerate():")
for index, name in enumerate(names):
    print(index, name)

print("\nUsing zip():")
marks = [85, 92, 78]
for name, mark in zip(names, marks):
    print(name, "scored", mark)

print("\nUsing sorted():")
for number in sorted([4, 1, 3, 2]):
    print(number)

print("\nUsing range():")
for number in range(2, 11, 2):
    print(number)
```
