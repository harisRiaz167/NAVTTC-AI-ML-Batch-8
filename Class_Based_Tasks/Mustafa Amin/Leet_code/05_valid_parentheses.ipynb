# Valid parentheses

A string of brackets `()`, `{}`, `[]` is valid when every opening bracket is closed by the same
type, in the correct order. A stack handles this: push openers, and on each closer the top of the
stack must be its matching opener.

> The original also checked whether the input appeared in a list of bracket pairs, which is a
> separate question from validity, so it is left out here. Characters other than brackets are ignored.

Standalone notebook: no imports or input prompts needed. Each `assert` raises an error if the solution is wrong, so no error means it works.

```python
MATCHING = {")": "(", "}": "{", "]": "["}


def is_valid(text: str) -> bool:
    stack = []
    for char in text:
        if char in MATCHING.values():
            stack.append(char)
        elif char in MATCHING:
            if not stack or stack.pop() != MATCHING[char]:
                return False
    return not stack


for sample in ["()", "()[]{}", "(]", "([)]", "{[]}", "((", ""]:
    print(f"{sample!r:>10} -> {is_valid(sample)}")

# Interactive version:
# print(is_valid(input("Enter a value: ")))

assert is_valid("()")
assert is_valid("()[]{}")
assert is_valid("{[]}")
assert is_valid("")
assert not is_valid("(]")
assert not is_valid("([)]")
assert not is_valid("((")
assert not is_valid(")")
```
