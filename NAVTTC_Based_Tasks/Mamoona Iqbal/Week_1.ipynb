**Assignment 8 — Week 1, Day 3**

**Topic: Python**

Write a Python program to find the area of a triangle using its three sides (the example uses a=5, b=6, c=7 and the semi-perimeter formula).


```python
import math

s1 = 7
s2 = 9
s3  = 12
# Calculate the semi-perimeter (half of the total distance around)
s = (s1 + s2 + s3) / 2
# Calculate the area using the formula
area_of_triangle = math.sqrt(s * (s - s1) * (s - s2) * (s - s3))

print("The area of the triangle is:", area_of_triangle)
```

Solve the quadratic equation ax² + bx + c = 0 using the cmath module (the example uses a=1, b=5, c=6).



quadratic.PNG

```python
import cmath
a = 2
b = 6
c = 9

# Calculate (b**2-4ac) (the part inside the square root)
d = (b**2) - (4 * a * c)
# use the quadratic formula
quad1 = (-b - cmath.sqrt(d)) / (2 * a)
quad2 = (-b + cmath.sqrt(d)) / (2 * a)

print("The quadratic equation solutions are:", quad1, "and", quad2)
```
