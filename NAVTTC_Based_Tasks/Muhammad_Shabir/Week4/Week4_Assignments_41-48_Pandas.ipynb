## Assignment 46 — Analyze data with Pandas

````python
import pandas as pd

df = pd.DataFrame({
    "Department": ["AI", "AI", "Data", "Data", "Web"],
    "Employee": ["Ali", "Sara", "Noor", "Hamza", "Zoya"],
    "Salary": [120000, 150000, 110000, 130000, 100000],
})
print("Mean salary:", df["Salary"].mean())
print("Maximum salary:", df["Salary"].max())
print("\nSalary by department:\n", df.groupby("Department")["Salary"].mean())
print("\nSorted by salary:\n", df.sort_values("Salary", ascending=False))
````

## Assignment 47 — Data cleaning

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ali", "Sara ", None, "Noor", "Ali"],
    "Age": [22, np.nan, 25, 21, 22],
    "Marks": [80, 95, 70, np.nan, 80],
})
print("Original:\n", df)

df["Name"] = df["Name"].str.strip()
df = df.drop_duplicates()
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["Name"] = df["Name"].fillna("Unknown")
print("\nCleaned:\n", df)
print("\nMissing values remaining:\n", df.isna().sum())
````

## Assignment 48 — Correlation

````python
import pandas as pd

df = pd.DataFrame({
    "StudyHours": [1, 2, 3, 4, 5],
    "PracticeTests": [1, 1, 2, 3, 4],
    "Marks": [45, 50, 60, 72, 88],
})
print(df)
print("\nCorrelation matrix:\n", df.corr(numeric_only=True))
````

### Correlation reminder

Correlation ranges from -1 to +1. A positive value means two variables tend to increase together; a negative value means one tends to decrease as the other increases. Correlation does not prove causation.
